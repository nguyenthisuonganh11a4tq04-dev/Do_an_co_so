## PhoBERT - Base

```
import os
import numpy as np
import torch
import torch.nn as nn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
from sklearn.metrics import (f1_score, confusion_matrix, accuracy_score,
                             recall_score, precision_score, roc_auc_score,
                             roc_curve, auc)
from sklearn.preprocessing import label_binarize
from transformers import (AutoTokenizer, AutoModelForSequenceClassification,
                          Trainer, TrainingArguments, EarlyStoppingCallback,
                          TrainerCallback, AutoConfig)
from torch.optim import AdamW
from transformers import get_linear_schedule_with_warmup
import re

# 0. Environment
os.environ["CUDA_LAUNCH_BLOCKING"] = "1"
os.environ["WANDB_DISABLED"] = "true"
os.environ["WANDB_MODE"] = "offline"

print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA device: {torch.cuda.get_device_name(0)}")

# 1. Label mapping  (string ↔ int)
LABEL2ID = {"negative": 0, "neutral": 1, "positive": 2}
ID2LABEL = {v: k for k, v in LABEL2ID.items()}
NUM_LABELS = len(LABEL2ID)

# 2. Load CSVs  (columns: sentence, sentiment)
DATA_DIR = "/content/drive/MyDrive/Do_an_co_so/DACS/data/dataset"

train_df = pd.read_csv(os.path.join(DATA_DIR, "train.csv")).dropna()
valid_df  = pd.read_csv(os.path.join(DATA_DIR, "valid.csv")).dropna()
test_df   = pd.read_csv(os.path.join(DATA_DIR, "test.csv")).dropna()

for df, name in [(train_df, "train"), (valid_df, "valid"), (test_df, "test")]:
    assert "sentence" in df.columns and "sentiment" in df.columns, \
        f"{name}.csv phải có cột 'sentence' và 'sentiment'"

train_y = train_df["sentiment"].map(LABEL2ID).values.astype(int)
valid_y  = valid_df["sentiment"].map(LABEL2ID).values.astype(int)
test_y   = test_df["sentiment"].map(LABEL2ID).values.astype(int)

print(f"\nLabel distribution – train : {np.bincount(train_y)}")
print(f"Label distribution – valid : {np.bincount(valid_y)}")
print(f"Label distribution – test  : {np.bincount(test_y)}")

assert all(0 <= y < NUM_LABELS for y in train_y), "train: nhãn ngoài [0,2]"
assert all(0 <= y < NUM_LABELS for y in valid_y),  "valid: nhãn ngoài [0,2]"
assert all(0 <= y < NUM_LABELS for y in test_y),   "test: nhãn ngoài [0,2]"

# 3. Load TEENCODE FILE — đọc đầy đủ 4 cột
teencode_path = "/content/drive/MyDrive/Do_an_co_so/DACS/data/teencode_final .xlsx"
teencode_df = pd.read_excel(teencode_path)
teencode_df = teencode_df.dropna(subset=[teencode_df.columns[0]]).drop_duplicates()

# Chuẩn hóa tên cột: strip khoảng trắng và lowercase
teencode_df.columns = [str(c).strip().lower() for c in teencode_df.columns]
print("Teencode columns (sau khi chuẩn hóa):", teencode_df.columns.tolist())

# ── Xác định 4 cột theo tên thực tế ──
COL_TEENCODE = "teencode"
COL_CHUAN    = "nghĩa"
COL_CAMXUC   = "nhãn cảm xúc"
COL_DOMANH   = "độ mạnh (1=nhẹ, 2=tb, 3=mạnh)"

# Kiểm tra cột tồn tại
required_cols = [COL_TEENCODE, COL_CHUAN, COL_CAMXUC, COL_DOMANH]
missing = [c for c in required_cols if c not in teencode_df.columns]
if missing:
    print(f"  Không tìm thấy cột: {missing}")
    print(f"   Các cột hiện có   : {teencode_df.columns.tolist()}")
    raise ValueError("Thiếu cột — copy tên cột từ dòng trên vào 4 biến COL_* rồi chạy lại.")

print(f"  Cột teencode   : '{COL_TEENCODE}'")
print(f"  Cột chuẩn      : '{COL_CHUAN}'")
print(f"  Cột cảm xúc    : '{COL_CAMXUC}'")
print(f"  Cột độ mạnh    : '{COL_DOMANH}'")

# Điền NaN
teencode_df[COL_CAMXUC] = teencode_df[COL_CAMXUC].fillna("").astype(str).str.strip()
teencode_df[COL_DOMANH] = teencode_df[COL_DOMANH].fillna("").astype(str).str.strip()

# ── Build 3 dict ──
teencode_norm_dict = dict(zip(
    teencode_df[COL_TEENCODE].astype(str).str.lower().str.strip(),
    teencode_df[COL_CHUAN].astype(str)
))

teencode_emotion_dict = dict(zip(
    teencode_df[COL_TEENCODE].astype(str).str.lower().str.strip(),
    teencode_df[COL_CAMXUC]
))

teencode_intensity_dict = dict(zip(
    teencode_df[COL_TEENCODE].astype(str).str.lower().str.strip(),
    teencode_df[COL_DOMANH]
))

# Cụm nhiều từ → ưu tiên thay trước (dài trước ngắn)
PHRASE_LIST = sorted(
    [(k, v) for k, v in teencode_norm_dict.items() if " " in k],
    key=lambda x: len(x[0]), reverse=True
)

print(f"\nSố lượng từ trong teencode dict: {len(teencode_norm_dict)}")
print("Sample (3 dòng đầu):")
for k in list(teencode_norm_dict.keys())[:3]:
    print(f"  '{k}' → nghĩa='{teencode_norm_dict[k]}' | "
          f"cảm xúc='{teencode_emotion_dict.get(k,'')}' | "
          f"độ mạnh='{teencode_intensity_dict.get(k,'')}' ")

# 4. Hàm tiền xử lý — dùng cả 4 cột
def normalize_teencode(text: str) -> str:
    """
    Bước 1: Lowercase
    Bước 2: Chuẩn hóa ký tự lặp ≥ 3 → giữ 2
    Bước 3: Thay cụm từ (phrase-level) + thu thập cảm xúc/độ mạnh
    Bước 4: Thay từng token + thu thập cảm xúc/độ mạnh
    Bước 5: Ghép tag [CẢM XÚC: ...] và [ĐỘ MẠNH: ...] vào cuối câu
    Bước 6: Xóa khoảng trắng thừa

    Ví dụ output:
        "hôm nay vui quá [CẢM XÚC: vui] [ĐỘ MẠNH: mạnh]"
    """
    if not isinstance(text, str):
        return ""

    emotions_found    = []
    intensities_found = []

    # Bước 1: Lowercase
    text = text.lower()

    # Bước 2: Chuẩn hóa ký tự lặp ≥ 3 → giữ 2
    text = re.sub(r"(.)\1{2,}", r"\1\1", text)

    # Bước 3: Thay cụm từ trước + thu thập meta
    for phrase, replacement in PHRASE_LIST:
        if phrase in text:
            emotion   = teencode_emotion_dict.get(phrase, "")
            intensity = teencode_intensity_dict.get(phrase, "")
            if emotion:
                emotions_found.append(emotion)
            if intensity:
                intensities_found.append(intensity)
            text = text.replace(phrase, f" {replacement} ")

    # Bước 4: Thay từng token + thu thập meta
    tokens = text.split()
    new_tokens = []
    for t in tokens:
        if t in teencode_norm_dict:
            emotion   = teencode_emotion_dict.get(t, "")
            intensity = teencode_intensity_dict.get(t, "")
            if emotion:
                emotions_found.append(emotion)
            if intensity:
                intensities_found.append(intensity)
            new_tokens.append(teencode_norm_dict[t])
        else:
            new_tokens.append(t)
    text = " ".join(new_tokens)

    # Bước 5: Ghép tag vào cuối câu
    if emotions_found:
        unique_emotions = list(dict.fromkeys(emotions_found))
        text += " [CẢM XÚC: " + ", ".join(unique_emotions) + "]"

    if intensities_found:
        unique_intensities = list(dict.fromkeys(intensities_found))
        text += " [ĐỘ MẠNH: " + ", ".join(unique_intensities) + "]"

    # Bước 6: Xóa khoảng trắng thừa
    text = re.sub(r"\s+", " ", text).strip()
    return text

# Áp dụng normalize lên 3 tập
print("\nĐang normalize train set...")
train_X = [normalize_teencode(x) for x in train_df["sentence"].tolist()]
print("Đang normalize valid set...")
valid_X  = [normalize_teencode(x) for x in valid_df["sentence"].tolist()]
print("Đang normalize test set...")
test_X   = [normalize_teencode(x) for x in test_df["sentence"].tolist()]

# Kiểm tra vài ví dụ để xác nhận
print("\nVí dụ normalize (5 câu đầu train):")
for i in range(min(5, len(train_X))):
    original = train_df["sentence"].tolist()[i]
    normalized = train_X[i]
    if original != normalized:
        print(f"  Gốc     : {original}")
        print(f"  Chuẩn   : {normalized}")
        print()


# 5. Tokenizer & Model
MODEL_NAME = "vinai/phobert-base-v2"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, use_fast=False)

config = AutoConfig.from_pretrained(
    MODEL_NAME,
    num_labels=NUM_LABELS,
    id2label=ID2LABEL,
    label2id=LABEL2ID,
)

# Tăng dropout để giảm overfitting
config.hidden_dropout_prob = 0.2
config.attention_probs_dropout_prob = 0.2

model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME,
    config=config,
    ignore_mismatched_sizes=True,
)

# 6. Tokenize
MAX_LENGTH = 128   # Có thể tăng lên 160 nếu các câu dài sau khi thêm tag

train_enc = tokenizer(train_X, truncation=True, padding=True, max_length=MAX_LENGTH)
valid_enc  = tokenizer(valid_X, truncation=True, padding=True, max_length=MAX_LENGTH)
test_enc   = tokenizer(test_X,  truncation=True, padding=True, max_length=MAX_LENGTH)

# 7. Dataset class
class SentimentDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {k: torch.tensor(v[idx]) for k, v in self.encodings.items()}
        item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

    def __len__(self):
        return len(self.labels)

train_dataset = SentimentDataset(train_enc, train_y)
valid_dataset  = SentimentDataset(valid_enc,  valid_y)
test_dataset   = SentimentDataset(test_enc,   test_y)

# 8. Label Smoothing Trainer
class LabelSmoothingTrainer(Trainer):
    def __init__(self, *args, label_smoothing=0.1, **kwargs):
        super().__init__(*args, **kwargs)
        self.label_smoothing = label_smoothing

    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits = outputs.logits
        loss_fct = nn.CrossEntropyLoss(label_smoothing=self.label_smoothing)
        loss = loss_fct(logits, labels)
        return (loss, outputs) if return_outputs else loss

# 9. LLRD Optimizer
def get_llrd_optimizer(model, base_lr=2e-5, decay_rate=0.9, weight_decay=0.01):
    """
    Layer-wise Learning Rate Decay:
      - Classifier head : base_lr * 2  (học nhanh nhất)
      - Layer 11 (top)  : base_lr * 0.9^1
      - Layer 10        : base_lr * 0.9^2
      ...
      - Embeddings      : base_lr * 0.9^13  (học chậm nhất)
    """
    encoder_layers = list(model.roberta.encoder.layer)
    num_layers = len(encoder_layers)
    param_groups = []

    # Classifier head
    param_groups.append({
        "params": model.classifier.parameters(),
        "lr": base_lr * 2,
        "weight_decay": weight_decay,
    })

    # Pooler
    if hasattr(model.roberta, "pooler") and model.roberta.pooler is not None:
        param_groups.append({
            "params": model.roberta.pooler.parameters(),
            "lr": base_lr,
            "weight_decay": weight_decay,
        })

    # Encoder layers (từ trên xuống dưới)
    for i, layer in enumerate(reversed(encoder_layers)):
        lr = base_lr * (decay_rate ** (i + 1))
        param_groups.append({
            "params": layer.parameters(),
            "lr": lr,
            "weight_decay": weight_decay,
        })

    # Embeddings
    param_groups.append({
        "params": model.roberta.embeddings.parameters(),
        "lr": base_lr * (decay_rate ** (num_layers + 1)),
        "weight_decay": weight_decay,
    })

    return AdamW(param_groups)

# 10. Compute metrics
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    return {
        "accuracy": accuracy_score(labels, preds),
        "f1_macro": f1_score(labels, preds, average="macro"),
    }

# 11. Epoch Logger Callback
class EpochLogger(TrainerCallback):
    def __init__(self):
        self.history = {
            "epoch": [],
            "train_loss": [],
            "val_loss": [],
            "val_acc": [],
            "val_f1": [],
        }
        self.step_losses = []
        self.steps = []
        self._last_train_loss = float("nan")

    def on_log(self, args, state, control, logs=None, **kwargs):
        if logs is None:
            return
        if "loss" in logs:
            self._last_train_loss = logs["loss"]
            self.step_losses.append(logs["loss"])
            self.steps.append(state.global_step)

        if "eval_loss" in logs:
            self.history["epoch"].append(round(state.epoch))
            self.history["train_loss"].append(self._last_train_loss)
            self.history["val_loss"].append(logs["eval_loss"])
            self.history["val_acc"].append(logs.get("eval_accuracy", float("nan")))
            self.history["val_f1"].append(logs.get("eval_f1_macro", float("nan")))

            print(f"\n[Epoch {state.epoch:.0f}]  "
                  f"train_loss={self._last_train_loss:.4f}  "
                  f"val_loss={logs['eval_loss']:.4f}  "
                  f"val_acc={logs.get('eval_accuracy', 0):.4f}  "
                  f"val_f1={logs.get('eval_f1_macro', 0):.4f}")

# 12. Training arguments
training_args = TrainingArguments(
    output_dir="./phobert_output",
    num_train_epochs=10,           # FIX: "1  0" → 10
    per_device_train_batch_size=16,
    per_device_eval_batch_size=32,
    learning_rate=2e-5,
    warmup_ratio=0.1,
    weight_decay=0.01,
    logging_dir="./logs",
    logging_steps=50,
    eval_strategy="epoch",
    save_strategy="epoch",
    save_total_limit=2,
    load_best_model_at_end=True,
    metric_for_best_model="f1_macro",
    greater_is_better=True,
    fp16=torch.cuda.is_available(),
    report_to="none",
    max_grad_norm=1.0,
)

# 13. Optimizer & Scheduler

optimizer = get_llrd_optimizer(model, base_lr=2e-5, decay_rate=0.9, weight_decay=0.01)

steps_per_epoch = len(train_dataset) // training_args.per_device_train_batch_size
total_steps     = steps_per_epoch * int(training_args.num_train_epochs)
warmup_steps    = int(total_steps * training_args.warmup_ratio)

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=warmup_steps,
    num_training_steps=total_steps,
)

# 14. Trainer
epoch_logger = EpochLogger()

trainer = LabelSmoothingTrainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=valid_dataset,
    compute_metrics=compute_metrics,
    optimizers=(optimizer, scheduler),
    callbacks=[
        epoch_logger,
        EarlyStoppingCallback(
            early_stopping_patience=3,
            early_stopping_threshold=0.001,
        ),
    ],
    label_smoothing=0.1,
)

# 15. Train & Evaluate
try:
    trainer.train()

    # 15a. Loss & F1 curves
    h = epoch_logger.history
    epochs = h["epoch"]

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    ax = axes[0]
    if epoch_logger.steps:
        ax.plot(epoch_logger.steps, epoch_logger.step_losses,
                alpha=0.3, color="steelblue", label="Train loss (step)")
    if epochs:
        ax.plot(epochs, h["train_loss"], marker="o", color="steelblue",
                linewidth=2, label="Train loss (epoch)")
        ax.plot(epochs, h["val_loss"],   marker="s", color="tomato",
                linewidth=2, linestyle="--", label="Val loss")
    ax.set_xlabel("Step / Epoch"); ax.set_ylabel("Loss")
    ax.set_title("Loss curves"); ax.legend(); ax.grid(True, alpha=0.3)

    ax = axes[1]
    if epochs:
        ax.plot(epochs, h["val_acc"], marker="o", color="mediumseagreen",
                linewidth=2, label="Val accuracy")
    ax.set_xlabel("Epoch"); ax.set_ylabel("Accuracy")
    ax.set_title("Validation Accuracy"); ax.legend(); ax.grid(True, alpha=0.3)

    ax = axes[2]
    if epochs:
        ax.plot(epochs, h["val_f1"], marker="o", color="mediumpurple",
                linewidth=2, label="Val F1 Macro")
    ax.set_xlabel("Epoch"); ax.set_ylabel("F1 Macro")
    ax.set_title("Validation F1 Macro"); ax.legend(); ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig("loss_accuracy_curves.png", dpi=150)
    plt.show()

    #15b. Evaluate on test set
    pred_out = trainer.predict(test_dataset)
    y_pred_probs = torch.nn.functional.softmax(
        torch.tensor(pred_out.predictions), dim=-1
    ).numpy()
    y_pred_ids = np.argmax(pred_out.predictions, axis=-1)
    y_true = test_y

    # Confusion matrix
    class_names = ["negative", "neutral", "positive"]
    cf = confusion_matrix(y_true, y_pred_ids)

    plt.figure(figsize=(7, 6))
    sn.heatmap(cf, annot=True, fmt="d", cmap="Blues",
               xticklabels=class_names, yticklabels=class_names,
               cbar_kws={"label": "Count"})
    plt.xlabel("Predicted", fontsize=12)
    plt.ylabel("True", fontsize=12)
    plt.title("Confusion Matrix – Test Set", fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig("confusion_matrix.png", dpi=150)
    plt.show()

    #Metrics
    acc        = accuracy_score(y_true, y_pred_ids)
    prec_macro = precision_score(y_true, y_pred_ids, average="macro")
    rec_macro  = recall_score(y_true, y_pred_ids, average="macro")
    f1_macro   = f1_score(y_true, y_pred_ids, average="macro")
    y_true_bin = label_binarize(y_true, classes=[0, 1, 2])
    auc_macro  = roc_auc_score(y_true_bin, y_pred_probs,
                               average="macro", multi_class="ovr")

    print("\n" + "=" * 60)
    print("FINAL TEST SET METRICS")
    print("=" * 60)
    print(f"{'Metric':<25} {'Value':>10}")
    print("-" * 60)
    print(f"{'Accuracy':<25} {acc:>10.4f}")
    print(f"{'Precision (macro)':<25} {prec_macro:>10.4f}")
    print(f"{'Recall (macro)':<25} {rec_macro:>10.4f}")
    print(f"{'F1-score (macro)':<25} {f1_macro:>10.4f}")
    print(f"{'AUC (macro)':<25} {auc_macro:>10.4f}")
    print("=" * 60)

    # Class-wise
    prec_cls = precision_score(y_true, y_pred_ids, average=None)
    rec_cls  = recall_score(y_true, y_pred_ids, average=None)
    f1_cls   = f1_score(y_true, y_pred_ids, average=None)

    print("\nCLASS-WISE METRICS")
    print("-" * 60)
    for i, name in enumerate(class_names):
        print(f"{name.upper():<12}  "
              f"P={prec_cls[i]:.4f}  R={rec_cls[i]:.4f}  F1={f1_cls[i]:.4f}")

    # Bar chart class-wise
    x = np.arange(len(class_names)); w = 0.25
    plt.figure(figsize=(9, 5))
    plt.bar(x - w, prec_cls, w, label="Precision", alpha=0.8)
    plt.bar(x,     rec_cls,  w, label="Recall",    alpha=0.8)
    plt.bar(x + w, f1_cls,   w, label="F1-score",  alpha=0.8)
    plt.xticks(x, class_names)
    plt.ylim([0, 1.1]); plt.ylabel("Score")
    plt.title("Class-wise Performance Metrics", fontweight="bold")
    plt.legend(); plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.savefig("classwise_metrics.png", dpi=150)
    plt.show()

    #ROC curves
    fpr, tpr, roc_auc_d = {}, {}, {}
    for i in range(NUM_LABELS):
        fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_pred_probs[:, i])
        roc_auc_d[i] = auc(fpr[i], tpr[i])
    fpr["macro"], tpr["macro"], _ = roc_curve(y_true_bin.ravel(), y_pred_probs.ravel())
    roc_auc_d["macro"] = auc(fpr["macro"], tpr["macro"])

    colors = ["#FF6B6B", "#4ECDC4", "#45B7D1"]
    plt.figure(figsize=(9, 7))
    for i in range(NUM_LABELS):
        plt.plot(fpr[i], tpr[i], color=colors[i], lw=2,
                 label=f"{class_names[i]} (AUC={roc_auc_d[i]:.4f})")
    plt.plot(fpr["macro"], tpr["macro"],
             color="deeppink", linestyle="--", lw=2.5,
             label=f"macro-avg (AUC={roc_auc_d['macro']:.4f})")
    plt.plot([0, 1], [0, 1], "k--", lw=1.5, alpha=0.5, label="Random")
    plt.xlim([0, 1]); plt.ylim([0, 1.05])
    plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
    plt.title("ROC Curves – Multi-class", fontweight="bold")
    plt.legend(loc="lower right"); plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig("roc_curves.png", dpi=150)
    plt.show()

    #Example predictions 
    print("\nEXAMPLE PREDICTIONS (first 10)")
    print("-" * 60)
    for i in range(min(10, len(test_X))):
        true_lbl = ID2LABEL[test_y[i]]
        pred_lbl = ID2LABEL[y_pred_ids[i]]
        probs    = y_pred_probs[i]
        mark     = "✓" if true_lbl == pred_lbl else "✗"
        print(f"[{mark}] Text: {test_X[i][:80]}")
        print(f"     True: {true_lbl:>8} | Pred: {pred_lbl:>8} | "
              f"neg={probs[0]:.3f} neu={probs[1]:.3f} pos={probs[2]:.3f}")
        print()

    all_results = {
        "PhoBERT-base-v2": {
            "accuracy":        acc,
            "precision_macro": prec_macro,
            "recall_macro":    rec_macro,
            "f1_macro":        f1_macro,
            "auc_macro":       auc_macro,
        }
    }
    print("\nStored in all_results:", all_results)

except RuntimeError as e:
    print(f"RuntimeError: {e}")
    if torch.cuda.is_available():
        print(f"GPU alloc: {torch.cuda.memory_allocated(0)/1e9:.2f} GB")
        print(f"GPU cache: {torch.cuda.memory_reserved(0)/1e9:.2f} GB")
        torch.cuda.empty_cache()
```

## SVM

```
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn

from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    confusion_matrix, roc_auc_score, roc_curve, auc,
    classification_report,
)

# 1. Load TEENCODE FILE — đọc đầy đủ 4 cột

teencode_path = "/content/drive/MyDrive/Do_an_co_so/DACS/data/teencode_final .xlsx"
teencode_df = pd.read_excel(teencode_path)

# Chuẩn hóa tên cột: strip khoảng trắng và lowercase
teencode_df.columns = [str(c).strip().lower() for c in teencode_df.columns]
print("Teencode columns (sau khi chuẩn hóa):", teencode_df.columns.tolist())

# ── Xác định 4 cột theo TÊN THỰC TẾ ──
COL_TEENCODE = "teencode"
COL_CHUAN    = "nghĩa"
COL_CAMXUC   = "nhãn cảm xúc"
COL_DOMANH   = "độ mạnh (1=nhẹ, 2=tb, 3=mạnh)"

# Kiểm tra tất cả cột tồn tại
required_cols = [COL_TEENCODE, COL_CHUAN, COL_CAMXUC, COL_DOMANH]
missing = [c for c in required_cols if c not in teencode_df.columns]
if missing:
    print(f"  Không tìm thấy cột: {missing}")
    print(f"   Các cột hiện có: {teencode_df.columns.tolist()}")
    raise ValueError("Thiếu cột — kiểm tra lại tên cột trong file Excel.")

print(f"  Cột teencode   : '{COL_TEENCODE}'")
print(f"  Cột chuẩn      : '{COL_CHUAN}'")
print(f"  Cột cảm xúc    : '{COL_CAMXUC}'")
print(f"  Cột độ mạnh    : '{COL_DOMANH}'")

# Lọc hàng có teencode và nghĩa hợp lệ
teencode_df = teencode_df.dropna(subset=[COL_TEENCODE, COL_CHUAN]).drop_duplicates()

# Điền NaN ở cảm xúc / độ mạnh
teencode_df[COL_CAMXUC] = teencode_df[COL_CAMXUC].fillna("").astype(str).str.strip()
teencode_df[COL_DOMANH] = teencode_df[COL_DOMANH].fillna("").astype(str).str.strip()

# ── Build 3 dict tra cứu theo key = teencode (lowercase) ──
# 1) Chuẩn hóa chính tả
teencode_norm_dict = {}
for _, row in teencode_df.iterrows():
    key = str(row[COL_TEENCODE]).strip().lower()
    val = str(row[COL_CHUAN]).strip().split("/")[0].strip()
    if key and val:
        teencode_norm_dict[key] = val

# 2) Cảm xúc
teencode_emotion_dict = dict(zip(
    teencode_df[COL_TEENCODE].astype(str).str.lower().str.strip(),
    teencode_df[COL_CAMXUC]
))

# 3) Độ mạnh
teencode_intensity_dict = dict(zip(
    teencode_df[COL_TEENCODE].astype(str).str.lower().str.strip(),
    teencode_df[COL_DOMANH]
))

print(f"\nSố lượng từ trong teencode dict: {len(teencode_norm_dict)}")
print("Sample (3 dòng đầu):")
for k in list(teencode_norm_dict.keys())[:3]:
    print(f"  '{k}' → nghĩa='{teencode_norm_dict[k]}' | "
          f"cảm xúc='{teencode_emotion_dict.get(k, '')}' | "
          f"độ mạnh='{teencode_intensity_dict.get(k, '')}'")

# 2. Hàm tiền xử lý — dùng cả 4 cột
def normalize_teencode(text: str) -> str:
    """
    Tiền xử lý văn bản tiếng Việt có chứa teencode:
    Bước 1: Lowercase
    Bước 2: Chuẩn hóa ký tự lặp ≥ 3 lần → giữ lại 2 (giữ cảm xúc)
    Bước 3: Thay thế cụm từ nhiều từ (phrase-level, ưu tiên cụm dài)
    Bước 4: Thay thế từng token đơn + thu thập cảm xúc & độ mạnh
    Bước 5: Ghép tag [CẢM XÚC: ...] và [ĐỘ MẠNH: ...] vào cuối câu
    Bước 6: Xóa khoảng trắng thừa

    Ví dụ output:
        "hôm nay vui quá [CẢM XÚC: vui] [ĐỘ MẠNH: mạnh]"
    """
    if not isinstance(text, str):
        return ""

    emotions_found    = []
    intensities_found = []

    # Bước 1: Lowercase
    text = text.lower()

    # Bước 2: Chuẩn hóa ký tự lặp ≥ 3 → giữ 2
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # Bước 3: Thay thế cụm từ (phrase) — ưu tiên cụm dài hơn
    sorted_phrases = sorted(
        [(k, v) for k, v in teencode_norm_dict.items() if ' ' in k],
        key=lambda x: len(x[0]),
        reverse=True
    )
    for phrase, replacement in sorted_phrases:
        if phrase in text:
            # Thu thập cảm xúc & độ mạnh của cụm từ
            emotion   = teencode_emotion_dict.get(phrase, "")
            intensity = teencode_intensity_dict.get(phrase, "")
            if emotion:
                emotions_found.append(emotion)
            if intensity:
                intensities_found.append(intensity)
            text = text.replace(phrase, f" {replacement} ")

    # Bước 4: Thay thế từng token đơn + thu thập meta
    tokens = text.split()
    new_tokens = []
    for t in tokens:
        if t in teencode_norm_dict:
            emotion   = teencode_emotion_dict.get(t, "")
            intensity = teencode_intensity_dict.get(t, "")
            if emotion:
                emotions_found.append(emotion)
            if intensity:
                intensities_found.append(intensity)
            new_tokens.append(teencode_norm_dict[t])
        else:
            new_tokens.append(t)
    text = " ".join(new_tokens)

    # Bước 5: Ghép tag cảm xúc & độ mạnh vào cuối câu
    if emotions_found:
        unique_emotions = list(dict.fromkeys(emotions_found))  # giữ thứ tự, bỏ trùng
        text += " [CẢM XÚC: " + ", ".join(unique_emotions) + "]"

    if intensities_found:
        unique_intensities = list(dict.fromkeys(intensities_found))
        text += " [ĐỘ MẠNH: " + ", ".join(unique_intensities) + "]"

    # Bước 6: Xóa khoảng trắng thừa
    text = re.sub(r'\s+', ' ', text).strip()

    return text

# 3. Load data

DATA_DIR = "/content/drive/MyDrive/Do_an_co_so/DACS/data/dataset"

train_df = pd.read_csv(os.path.join(DATA_DIR, "train.csv")).dropna()
valid_df  = pd.read_csv(os.path.join(DATA_DIR, "valid.csv")).dropna()
test_df   = pd.read_csv(os.path.join(DATA_DIR, "test.csv")).dropna()

# Gộp train + valid (SVM không cần validation set riêng)
train_full_df = pd.concat([train_df, valid_df], ignore_index=True)

# Áp dụng tiền xử lý
print("\nĐang tiền xử lý văn bản...")
X_train = [normalize_teencode(s) for s in train_full_df["sentence"].tolist()]
X_test  = [normalize_teencode(s) for s in test_df["sentence"].tolist()]

# Encode nhãn
le = LabelEncoder()
le.classes_ = np.array(["negative", "neutral", "positive"])
y_train = le.transform(train_full_df["sentiment"].tolist())
y_test  = le.transform(test_df["sentiment"].tolist())

CLASS_NAMES = le.classes_.tolist()
print(f"\nClasses: {CLASS_NAMES}")
print(f"Train size: {len(X_train)} | Test size: {len(X_test)}")
print(f"Train label dist: {np.bincount(y_train)}")
print(f"Test  label dist: {np.bincount(y_test)}")

# Kiểm tra vài mẫu sau khi chuẩn hóa
print("\nVí dụ normalize (5 câu đầu train):")
for i in range(min(5, len(X_train))):
    orig = train_full_df["sentence"].iloc[i]
    norm = X_train[i]
    if orig.lower() != norm:
        print(f"  Gốc  : {orig}")
        print(f"  Chuẩn: {norm}")
        print()


# 4. Pipeline: TF-IDF + SVM

pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(
        ngram_range=(1, 3),      # unigram + bigram + trigram
        max_features=60_000,
        sublinear_tf=True,       # log(TF) — giảm ảnh hưởng từ lặp nhiều
        analyzer="word",
        min_df=1,                # giữ cả từ xuất hiện 1 lần (teencode hiếm)
        token_pattern=r'\S+',   # giữ token ngắn như "k", "dc", "vl"
                                 # và giữ token dạng "[CẢM XÚC: ...]"
    )),
    ("svm", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
        class_weight="balanced",
        probability=True,        # cần để predict_proba → ROC/AUC
        random_state=42,
    )),
])

# 5. Train

print("\nĐang huấn luyện SVM...")
pipeline.fit(X_train, y_train)
print("Huấn luyện xong.")


# 6. Predict

y_pred       = pipeline.predict(X_test)
y_pred_probs = pipeline.predict_proba(X_test)   # shape (N, 3)


# 7. Metrics

acc        = accuracy_score(y_test, y_pred)
prec_macro = precision_score(y_test, y_pred, average="macro")
rec_macro  = recall_score(y_test, y_pred, average="macro")
f1_macro   = f1_score(y_test, y_pred, average="macro")
y_test_bin = label_binarize(y_test, classes=[0, 1, 2])
auc_macro  = roc_auc_score(y_test_bin, y_pred_probs,
                            average="macro", multi_class="ovr")

print("\n" + "=" * 60)
print("FINAL TEST SET METRICS")
print("=" * 60)
print(f"{'Metric':<25} {'Value':>10}")
print("-" * 60)
print(f"{'Accuracy':<25} {acc:>10.4f}")
print(f"{'Precision (macro)':<25} {prec_macro:>10.4f}")
print(f"{'Recall (macro)':<25} {rec_macro:>10.4f}")
print(f"{'F1-score (macro)':<25} {f1_macro:>10.4f}")
print(f"{'AUC (macro)':<25} {auc_macro:>10.4f}")
print("=" * 60)

print("\nCLASSIFICATION REPORT")
print(classification_report(y_test, y_pred, target_names=CLASS_NAMES))


# 8. Confusion Matrix

cf = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(7, 6))
sn.heatmap(cf, annot=True, fmt="d", cmap="Blues",
           xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
           cbar_kws={"label": "Count"})
plt.xlabel("Predicted", fontsize=12)
plt.ylabel("True", fontsize=12)
plt.title("Confusion Matrix – SVM + Teencode (Test Set)", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("svm_confusion_matrix.png", dpi=150)
plt.show()


# 9. Class-wise bar chart

prec_cls = precision_score(y_test, y_pred, average=None)
rec_cls  = recall_score(y_test, y_pred, average=None)
f1_cls   = f1_score(y_test, y_pred, average=None)

x = np.arange(len(CLASS_NAMES)); w = 0.25
plt.figure(figsize=(9, 5))
plt.bar(x - w, prec_cls, w, label="Precision", alpha=0.85)
plt.bar(x,     rec_cls,  w, label="Recall",    alpha=0.85)
plt.bar(x + w, f1_cls,   w, label="F1-score",  alpha=0.85)
plt.xticks(x, CLASS_NAMES)
plt.ylim([0, 1.1]); plt.ylabel("Score")
plt.title("Class-wise Performance Metrics – SVM + Teencode", fontweight="bold")
plt.legend(); plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("svm_classwise_metrics.png", dpi=150)
plt.show()

# 10. ROC Curves

fpr, tpr, roc_auc_d = {}, {}, {}
for i in range(3):
    fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, i], y_pred_probs[:, i])
    roc_auc_d[i] = auc(fpr[i], tpr[i])
fpr["macro"], tpr["macro"], _ = roc_curve(y_test_bin.ravel(), y_pred_probs.ravel())
roc_auc_d["macro"] = auc(fpr["macro"], tpr["macro"])

colors = ["#FF6B6B", "#4ECDC4", "#45B7D1"]
plt.figure(figsize=(9, 7))
for i in range(3):
    plt.plot(fpr[i], tpr[i], color=colors[i], lw=2,
             label=f"{CLASS_NAMES[i]} (AUC={roc_auc_d[i]:.4f})")
plt.plot(fpr["macro"], tpr["macro"],
         color="deeppink", linestyle="--", lw=2.5,
         label=f"macro-avg (AUC={roc_auc_d['macro']:.4f})")
plt.plot([0, 1], [0, 1], "k--", lw=1.5, alpha=0.5, label="Random")
plt.xlim([0, 1]); plt.ylim([0, 1.05])
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curves – SVM + Teencode", fontweight="bold")
plt.legend(loc="lower right"); plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("svm_roc_curves.png", dpi=150)
plt.show()


# 11. Example predictions

print("\nEXAMPLE PREDICTIONS (first 10 test samples)")
print("-" * 70)
for i in range(min(10, len(X_test))):
    orig_text = test_df["sentence"].iloc[i]
    norm_text = X_test[i]
    true_lbl  = le.inverse_transform([y_test[i]])[0]
    pred_lbl  = le.inverse_transform([y_pred[i]])[0]
    probs     = y_pred_probs[i]
    mark      = "✓" if true_lbl == pred_lbl else "✗"

    print(f"[{mark}] Gốc  : {orig_text[:70]}")
    if orig_text.lower().strip() != norm_text:
        print(f"     Chuẩn: {norm_text[:70]}")
    print(f"     True: {true_lbl:>8} | Pred: {pred_lbl:>8} | "
          f"neg={probs[0]:.3f} neu={probs[1]:.3f} pos={probs[2]:.3f}")
    print()

# Lưu kết quả

all_results = {}
all_results["SVM_teencode"] = {
    "accuracy":        acc,
    "precision_macro": prec_macro,
    "recall_macro":    rec_macro,
    "f1_macro":        f1_macro,
    "auc_macro":       auc_macro,
}
print("Stored in all_results:", all_results)
```

```
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn

import torch
import torch.nn as nn
from torch.utils.data import Dataset

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    Trainer,
    TrainingArguments,
    EarlyStoppingCallback,
    TrainerCallback,
)
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    confusion_matrix, roc_auc_score, roc_curve, auc,
    classification_report,
)
from sklearn.preprocessing import label_binarize

# 0. Environment
os.environ["WANDB_DISABLED"]       = "true"
os.environ["WANDB_MODE"]           = "offline"
os.environ["CUDA_LAUNCH_BLOCKING"] = "1"

print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")

# 1. Config
class Config:
    MODEL_NAME    = "xlm-roberta-base"
    TEENCODE_PATH = "/content/drive/MyDrive/Do_an_co_so/DACS/data/teencode_final .xlsx"
    DATA_DIR      = "/content/drive/MyDrive/Do_an_co_so/DACS/data/dataset"
    MAX_LENGTH    = 128
    BATCH_SIZE    = 16
    EPOCHS        = 5
    LR            = 2e-5
    WARMUP_RATIO  = 0.1
    WEIGHT_DECAY  = 0.01
    DROPOUT       = 0.1
    SEED          = 42
    OUTPUT_DIR    = "./xlmr_output"
    FP16          = torch.cuda.is_available()

cfg = Config()

LABEL2ID    = {"negative": 0, "neutral": 1, "positive": 2}
ID2LABEL    = {v: k for k, v in LABEL2ID.items()}
CLASS_NAMES = ["negative", "neutral", "positive"]

# 2. Load TEENCODE FILE — đọc đầy đủ 4 cột

teencode_df = pd.read_excel(cfg.TEENCODE_PATH)

# Chuẩn hóa tên cột: strip khoảng trắng và lowercase
teencode_df.columns = [str(c).strip().lower() for c in teencode_df.columns]
print("Teencode columns (sau khi chuẩn hóa):", teencode_df.columns.tolist())

# ── Xác định 4 cột theo tên thực tế ──
COL_TEENCODE = "teencode"
COL_CHUAN    = "nghĩa"
COL_CAMXUC   = "nhãn cảm xúc"
COL_DOMANH   = "độ mạnh (1=nhẹ, 2=tb, 3=mạnh)"

# Kiểm tra cột tồn tại
required_cols = [COL_TEENCODE, COL_CHUAN, COL_CAMXUC, COL_DOMANH]
missing = [c for c in required_cols if c not in teencode_df.columns]
if missing:
    print(f"  Không tìm thấy cột: {missing}")
    print(f"   Các cột hiện có   : {teencode_df.columns.tolist()}")
    raise ValueError("Thiếu cột — copy tên cột từ dòng trên vào 4 biến COL_* rồi chạy lại.")

print(f"  Cột teencode   : '{COL_TEENCODE}'")
print(f"  Cột chuẩn      : '{COL_CHUAN}'")
print(f"  Cột cảm xúc    : '{COL_CAMXUC}'")
print(f"  Cột độ mạnh    : '{COL_DOMANH}'")

# Lọc hàng có teencode hợp lệ
teencode_df = teencode_df.dropna(subset=[COL_TEENCODE, COL_CHUAN]).drop_duplicates()

# Điền NaN ở cảm xúc / độ mạnh bằng chuỗi rỗng
teencode_df[COL_CAMXUC] = teencode_df[COL_CAMXUC].fillna("").astype(str).str.strip()
teencode_df[COL_DOMANH] = teencode_df[COL_DOMANH].fillna("").astype(str).str.strip()

# ── Build 3 dict tra cứu ──
teencode_norm_dict = {}
for _, row in teencode_df.iterrows():
    key = str(row[COL_TEENCODE]).strip().lower()
    val = str(row[COL_CHUAN]).strip().split("/")[0].strip()
    if key and val:
        teencode_norm_dict[key] = val

teencode_emotion_dict = dict(zip(
    teencode_df[COL_TEENCODE].astype(str).str.lower().str.strip(),
    teencode_df[COL_CAMXUC]
))

teencode_intensity_dict = dict(zip(
    teencode_df[COL_TEENCODE].astype(str).str.lower().str.strip(),
    teencode_df[COL_DOMANH]
))

# Cụm nhiều từ → ưu tiên thay trước (dài trước ngắn)
PHRASE_LIST = sorted(
    [(k, v) for k, v in teencode_norm_dict.items() if " " in k],
    key=lambda x: len(x[0]), reverse=True
)

print(f"\nSố lượng từ trong teencode dict: {len(teencode_norm_dict)}")
print("Sample (3 dòng đầu):")
for k in list(teencode_norm_dict.keys())[:3]:
    print(f"  '{k}' → nghĩa='{teencode_norm_dict[k]}' | "
          f"cảm xúc='{teencode_emotion_dict.get(k, '')}' | "
          f"độ mạnh='{teencode_intensity_dict.get(k, '')}'")

# 3. Hàm tiền xử lý — dùng cả 4 cột
def normalize_teencode(text: str) -> str:
    """
    Bước 1: Lowercase
    Bước 2: Chuẩn hóa ký tự lặp ≥ 3 → giữ 2
    Bước 3: Thay cụm từ (phrase-level) + thu thập cảm xúc/độ mạnh
    Bước 4: Thay từng token + thu thập cảm xúc/độ mạnh
    Bước 5: Ghép tag [CẢM XÚC: ...] và [ĐỘ MẠNH: ...] vào cuối câu
    Bước 6: Xóa khoảng trắng thừa

    Ví dụ output:
        "hôm nay vui quá [CẢM XÚC: vui] [ĐỘ MẠNH: mạnh]"
    """
    if not isinstance(text, str):
        return ""

    emotions_found    = []
    intensities_found = []

    # Bước 1: Lowercase
    text = text.lower()

    # Bước 2: Chuẩn hóa ký tự lặp ≥ 3 → giữ 2
    text = re.sub(r"(.)\1{2,}", r"\1\1", text)

    # Bước 3: Thay cụm từ trước + thu thập meta
    for phrase, replacement in PHRASE_LIST:
        if phrase in text:
            emotion   = teencode_emotion_dict.get(phrase, "")
            intensity = teencode_intensity_dict.get(phrase, "")
            if emotion:
                emotions_found.append(emotion)
            if intensity:
                intensities_found.append(intensity)
            text = text.replace(phrase, f" {replacement} ")

    # Bước 4: Thay từng token + thu thập meta
    tokens = text.split()
    new_tokens = []
    for t in tokens:
        if t in teencode_norm_dict:
            emotion   = teencode_emotion_dict.get(t, "")
            intensity = teencode_intensity_dict.get(t, "")
            if emotion:
                emotions_found.append(emotion)
            if intensity:
                intensities_found.append(intensity)
            new_tokens.append(teencode_norm_dict[t])
        else:
            new_tokens.append(t)
    text = " ".join(new_tokens)

    # Bước 5: Ghép tag vào cuối câu
    if emotions_found:
        unique_emotions = list(dict.fromkeys(emotions_found))
        text += " [CẢM XÚC: " + ", ".join(unique_emotions) + "]"

    if intensities_found:
        unique_intensities = list(dict.fromkeys(intensities_found))
        text += " [ĐỘ MẠNH: " + ", ".join(unique_intensities) + "]"

    # Bước 6: Xóa khoảng trắng thừa
    text = re.sub(r"\s+", " ", text).strip()
    return text

# 4. Load & preprocess data
train_df = pd.read_csv(os.path.join(cfg.DATA_DIR, "train.csv")).dropna()
valid_df  = pd.read_csv(os.path.join(cfg.DATA_DIR, "valid.csv")).dropna()
test_df   = pd.read_csv(os.path.join(cfg.DATA_DIR, "test.csv")).dropna()

print("\nĐang tiền xử lý teencode...")
train_X = [normalize_teencode(s) for s in train_df["sentence"]]
valid_X  = [normalize_teencode(s) for s in valid_df["sentence"]]
test_X   = [normalize_teencode(s) for s in test_df["sentence"]]

# Kiểm tra vài mẫu sau normalize
print("\nVí dụ normalize (5 câu đầu train):")
for i in range(min(5, len(train_X))):
    orig = train_df["sentence"].iloc[i]
    norm = train_X[i]
    if orig.lower().strip() != norm:
        print(f"  Gốc  : {orig}")
        print(f"  Chuẩn: {norm}")
        print()

train_y = train_df["sentiment"].map(LABEL2ID).values.astype(int)
valid_y  = valid_df["sentiment"].map(LABEL2ID).values.astype(int)
test_y   = test_df["sentiment"].map(LABEL2ID).values.astype(int)

print(f"Train: {len(train_X)} | Valid: {len(valid_X)} | Test: {len(test_X)}")
print(f"Label dist train : {np.bincount(train_y)}")
print(f"Label dist valid : {np.bincount(valid_y)}")
print(f"Label dist test  : {np.bincount(test_y)}")

# 5. Tokenizer
tokenizer = AutoTokenizer.from_pretrained(cfg.MODEL_NAME)

print("\nTokenizing...")
train_enc = tokenizer(train_X, truncation=True, padding=True,
                      max_length=cfg.MAX_LENGTH)
valid_enc  = tokenizer(valid_X,  truncation=True, padding=True,
                       max_length=cfg.MAX_LENGTH)
test_enc   = tokenizer(test_X,   truncation=True, padding=True,
                       max_length=cfg.MAX_LENGTH)
print("Tokenizing done.")

# 6. Dataset class
class SentimentDataset(Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels    = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        item = {k: torch.tensor(v[idx]) for k, v in self.encodings.items()}
        item["labels"] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

train_dataset = SentimentDataset(train_enc, train_y)
valid_dataset  = SentimentDataset(valid_enc,  valid_y)
test_dataset   = SentimentDataset(test_enc,   test_y)

# 7. Model
model = AutoModelForSequenceClassification.from_pretrained(
    cfg.MODEL_NAME,
    num_labels         = len(LABEL2ID),
    id2label           = ID2LABEL,
    label2id           = LABEL2ID,
    classifier_dropout = cfg.DROPOUT,
)

# 8. Class-weighted loss
class_counts  = np.bincount(train_y)
class_weights = torch.tensor(1.0 / class_counts, dtype=torch.float)
class_weights = class_weights / class_weights.sum() * len(LABEL2ID)

class WeightedTrainer(Trainer):
    def __init__(self, class_weights, **kwargs):
        super().__init__(**kwargs)
        self.class_weights = class_weights.to(self.args.device)

    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels  = inputs.pop("labels")
        outputs = model(**inputs)
        logits  = outputs.logits
        loss    = nn.CrossEntropyLoss(weight=self.class_weights)(logits, labels)
        return (loss, outputs) if return_outputs else loss

# 9. Compute metrics
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    f1_per_class = f1_score(labels, preds, average=None)
    return {
        "accuracy": accuracy_score(labels, preds),
        "f1_macro": f1_score(labels, preds, average="macro"),
        "f1_neg":   f1_per_class[0],
        "f1_neu":   f1_per_class[1],
        "f1_pos":   f1_per_class[2],
    }

# 10. Epoch Logger Callback
#     [FIX] Không gọi predict() 2 lần trong on_log
#     — dùng lại logs từ eval đã có, chỉ gọi predict test 1 lần/epoch
class EpochLogger(TrainerCallback):
    def __init__(self, test_dataset):
        self.test_dataset = test_dataset
        self._trainer     = None
        self.history = {
            "epoch":      [],
            "train_loss": [],
            "val_loss":   [],
            "val_acc":    [],
            "val_f1":     [],
            "test_loss":  [],
            "test_acc":   [],
            "test_f1":    [],
        }
        self.step_losses      = []
        self.steps            = []
        self._last_train_loss = float("nan")

    def set_trainer(self, trainer):
        self._trainer = trainer

    def on_log(self, args, state, control, logs=None, **kwargs):
        if logs is None:
            return

        # Ghi step-level train loss
        if "loss" in logs:
            self._last_train_loss = logs["loss"]
            self.step_losses.append(logs["loss"])
            self.steps.append(state.global_step)

        # Ghi epoch-level — chỉ khi có eval_loss
        if "eval_loss" in logs and self._trainer is not None:
            # Val metrics — dùng lại từ logs (đã có sẵn, không gọi predict lại)
            val_acc = logs.get("eval_accuracy", float("nan"))
            val_f1  = logs.get("eval_f1_macro", float("nan"))

            # Test — gọi predict 1 lần/epoch
            test_out  = self._trainer.predict(self.test_dataset)
            test_pred = np.argmax(test_out.predictions, axis=-1)
            test_acc  = accuracy_score(test_out.label_ids, test_pred)
            test_f1   = f1_score(test_out.label_ids, test_pred, average="macro")
            test_loss = test_out.metrics.get("test_loss", float("nan"))

            self.history["epoch"].append(round(state.epoch))
            self.history["train_loss"].append(self._last_train_loss)
            self.history["val_loss"].append(logs["eval_loss"])
            self.history["val_acc"].append(val_acc)
            self.history["val_f1"].append(val_f1)
            self.history["test_loss"].append(test_loss)
            self.history["test_acc"].append(test_acc)
            self.history["test_f1"].append(test_f1)

            print(f"\n[Epoch {state.epoch:.0f}]  "
                  f"train_loss={self._last_train_loss:.4f}  "
                  f"val_loss={logs['eval_loss']:.4f}  val_f1={val_f1:.4f}  "
                  f"test_acc={test_acc:.4f}  test_f1={test_f1:.4f}")

# 11. Training arguments
training_args = TrainingArguments(
    output_dir                  = cfg.OUTPUT_DIR,
    num_train_epochs            = cfg.EPOCHS,
    per_device_train_batch_size = cfg.BATCH_SIZE,
    per_device_eval_batch_size  = cfg.BATCH_SIZE * 2,
    learning_rate               = cfg.LR,
    warmup_ratio                = cfg.WARMUP_RATIO,
    weight_decay                = cfg.WEIGHT_DECAY,
    logging_dir                 = "./logs",
    logging_steps               = 50,           # mịn hơn để EpochLogger có step loss đẹp
    eval_strategy               = "epoch",
    save_strategy               = "epoch",
    save_total_limit            = 2,
    load_best_model_at_end      = True,
    metric_for_best_model       = "f1_macro",
    greater_is_better           = True,
    fp16                        = cfg.FP16,
    report_to                   = "none",
    seed                        = cfg.SEED,
    dataloader_num_workers      = 0,
    max_grad_norm               = 1.0,          # gradient clipping
)

# 12. Trainer
epoch_logger = EpochLogger(test_dataset)

trainer = WeightedTrainer(
    class_weights   = class_weights,
    model           = model,
    args            = training_args,
    train_dataset   = train_dataset,
    eval_dataset    = valid_dataset,
    compute_metrics = compute_metrics,
    callbacks       = [
        epoch_logger,
        EarlyStoppingCallback(
            early_stopping_patience=2,
            early_stopping_threshold=0.001,
        ),
    ],
)

epoch_logger.set_trainer(trainer)

# 13. Train & Evaluate
try:
    print("\nBắt đầu huấn luyện XLM-RoBERTa...")
    trainer.train()

    #Plot training curves 
    h      = epoch_logger.history
    epochs = h["epoch"]

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    # Loss
    if epoch_logger.steps:
        axes[0].plot(epoch_logger.steps, epoch_logger.step_losses,
                     alpha=0.4, color="steelblue", label="Train loss (step)")
    if epochs:
        axes[0].plot(epochs, h["train_loss"], marker="o", color="steelblue",
                     linewidth=2, label="Train loss (epoch)")
        axes[0].plot(epochs, h["val_loss"],  marker="s", color="tomato",
                     linewidth=2, linestyle="--", label="Val loss")
        axes[0].plot(epochs, h["test_loss"], marker="^", color="orange",
                     linewidth=2, linestyle=":", label="Test loss")
    axes[0].set_xlabel("Step / Epoch"); axes[0].set_ylabel("Loss")
    axes[0].set_title("Loss Curves"); axes[0].legend(); axes[0].grid(True, alpha=0.3)

    # Accuracy
    if epochs:
        axes[1].plot(epochs, h["val_acc"],  marker="o", color="mediumseagreen",
                     linewidth=2, label="Val Accuracy")
        axes[1].plot(epochs, h["test_acc"], marker="s", color="darkorange",
                     linewidth=2, linestyle="--", label="Test Accuracy")
    axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Accuracy")
    axes[1].set_title("Accuracy per Epoch")
    axes[1].legend(); axes[1].grid(True, alpha=0.3)

    # F1
    if epochs:
        axes[2].plot(epochs, h["val_f1"],  marker="o", color="mediumpurple",
                     linewidth=2, label="Val F1 macro")
        axes[2].plot(epochs, h["test_f1"], marker="s", color="crimson",
                     linewidth=2, linestyle="--", label="Test F1 macro")
    axes[2].set_xlabel("Epoch"); axes[2].set_ylabel("F1 Macro")
    axes[2].set_title("F1 Macro per Epoch")
    axes[2].legend(); axes[2].grid(True, alpha=0.3)

    plt.suptitle("XLM-RoBERTa – Training History", fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig("xlmr_training_curves.png", dpi=150)
    plt.show()

    #Evaluate on test set 
    pred_out     = trainer.predict(test_dataset)
    y_pred_probs = torch.softmax(
        torch.tensor(pred_out.predictions), dim=-1
    ).numpy()
    y_pred = np.argmax(pred_out.predictions, axis=-1)
    y_true = test_y

    #Metrics
    acc        = accuracy_score(y_true, y_pred)
    prec_macro = precision_score(y_true, y_pred, average="macro")
    rec_macro  = recall_score(y_true, y_pred, average="macro")
    f1_macro   = f1_score(y_true, y_pred, average="macro")
    y_true_bin = label_binarize(y_true, classes=[0, 1, 2])
    auc_macro  = roc_auc_score(y_true_bin, y_pred_probs,
                                average="macro", multi_class="ovr")

    print("\n" + "=" * 60)
    print("FINAL TEST SET METRICS – XLM-RoBERTa + Teencode")
    print("=" * 60)
    print(f"{'Metric':<25} {'Value':>10}")
    print("-" * 60)
    print(f"{'Accuracy':<25} {acc:>10.4f}")
    print(f"{'Precision (macro)':<25} {prec_macro:>10.4f}")
    print(f"{'Recall (macro)':<25} {rec_macro:>10.4f}")
    print(f"{'F1-score (macro)':<25} {f1_macro:>10.4f}")
    print(f"{'AUC (macro)':<25} {auc_macro:>10.4f}")
    print("=" * 60)

    print("\nCLASSIFICATION REPORT")
    print(classification_report(y_true, y_pred, target_names=CLASS_NAMES))

    #Confusion Matrix
    cf = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(7, 6))
    sn.heatmap(cf, annot=True, fmt="d", cmap="Blues",
               xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
               cbar_kws={"label": "Count"})
    plt.xlabel("Predicted", fontsize=12)
    plt.ylabel("True", fontsize=12)
    plt.title("Confusion Matrix – XLM-RoBERTa (Test Set)",
              fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig("xlmr_confusion_matrix.png", dpi=150)
    plt.show()

    #Class-wise bar chart
    prec_cls = precision_score(y_true, y_pred, average=None)
    rec_cls  = recall_score(y_true, y_pred, average=None)
    f1_cls   = f1_score(y_true, y_pred, average=None)

    x = np.arange(len(CLASS_NAMES)); w = 0.25
    plt.figure(figsize=(9, 5))
    plt.bar(x - w, prec_cls, w, label="Precision", alpha=0.85)
    plt.bar(x,     rec_cls,  w, label="Recall",    alpha=0.85)
    plt.bar(x + w, f1_cls,   w, label="F1-score",  alpha=0.85)
    plt.xticks(x, CLASS_NAMES)
    plt.ylim([0, 1.1]); plt.ylabel("Score")
    plt.title("Class-wise Metrics – XLM-RoBERTa + Teencode", fontweight="bold")
    plt.legend(); plt.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.savefig("xlmr_classwise_metrics.png", dpi=150)
    plt.show()

    #ROC Curves
    fpr, tpr, roc_auc_d = {}, {}, {}
    for i in range(3):
        fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_pred_probs[:, i])
        roc_auc_d[i]      = auc(fpr[i], tpr[i])
    fpr["macro"], tpr["macro"], _ = roc_curve(
        y_true_bin.ravel(), y_pred_probs.ravel()
    )
    roc_auc_d["macro"] = auc(fpr["macro"], tpr["macro"])

    colors = ["#FF6B6B", "#4ECDC4", "#45B7D1"]
    plt.figure(figsize=(9, 7))
    for i in range(3):
        plt.plot(fpr[i], tpr[i], color=colors[i], lw=2,
                 label=f"{CLASS_NAMES[i]} (AUC={roc_auc_d[i]:.4f})")
    plt.plot(fpr["macro"], tpr["macro"],
             color="deeppink", linestyle="--", lw=2.5,
             label=f"macro-avg (AUC={roc_auc_d['macro']:.4f})")
    plt.plot([0, 1], [0, 1], "k--", lw=1.5, alpha=0.5, label="Random")
    plt.xlim([0, 1]); plt.ylim([0, 1.05])
    plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
    plt.title("ROC Curves – XLM-RoBERTa + Teencode", fontweight="bold")
    plt.legend(loc="lower right"); plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig("xlmr_roc_curves.png", dpi=150)
    plt.show()

    #Example predictions
    print("\nEXAMPLE PREDICTIONS (first 10 test samples)")
    print("-" * 70)
    for i in range(min(10, len(test_X))):
        orig     = test_df["sentence"].iloc[i]
        norm     = test_X[i]
        true_lbl = ID2LABEL[y_true[i]]
        pred_lbl = ID2LABEL[y_pred[i]]
        probs    = y_pred_probs[i]
        mark     = "✓" if true_lbl == pred_lbl else "✗"
        print(f"[{mark}] Gốc  : {orig[:70]}")
        if orig.lower().strip() != norm:
            print(f"     Chuẩn: {norm[:70]}")
        print(f"     True: {true_lbl:>8} | Pred: {pred_lbl:>8} | "
              f"neg={probs[0]:.3f} neu={probs[1]:.3f} pos={probs[2]:.3f}")
        print()

    # Lưu kết quả
    all_results = {}
    all_results["XLM-RoBERTa"] = {
        "accuracy":        acc,
        "precision_macro": prec_macro,
        "recall_macro":    rec_macro,
        "f1_macro":        f1_macro,
        "auc_macro":       auc_macro,
    }
    print("Stored in all_results:", all_results)

except RuntimeError as e:
    print(f"RuntimeError: {e}")
    if torch.cuda.is_available():
        print(f"GPU alloc : {torch.cuda.memory_allocated(0)/1e9:.2f} GB")
        print(f"GPU cache : {torch.cuda.memory_reserved(0)/1e9:.2f} GB")
        torch.cuda.empty_cache()
```

## CNN Text

```
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.optim import Adam
from torch.optim.lr_scheduler import ReduceLROnPlateau
from collections import Counter

from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    confusion_matrix, roc_auc_score, roc_curve, auc,
    classification_report,
)
from sklearn.preprocessing import label_binarize


# 0. Config
class Config:
    DATA_DIR        = "/content/drive/MyDrive/Do_an_co_so/DACS/data/dataset"
    TEENCODE_PATH   = "/content/drive/MyDrive/Do_an_co_so/DACS/data/teencode_final .xlsx"
    SEED            = 42
    MAX_LENGTH      = 64          # P95 = 47 từ → 64 cho margin
    VOCAB_SIZE      = 20_000      # top N từ phổ biến nhất
    EMBED_DIM       = 128         # embedding dimension
    NUM_FILTERS     = 128         # số filter mỗi kernel
    KERNEL_SIZES    = [2, 3, 4, 5]
    DROPOUT         = 0.5
    NUM_CLASSES     = 3
    BATCH_SIZE      = 64
    EPOCHS          = 20
    LR              = 1e-3
    PATIENCE        = 4           # early stopping
    DEVICE          = "cuda" if torch.cuda.is_available() else "cpu"

    #Cột trong file teencode (4-cột mới) 
    COL_TEENCODE    = "Teencode"           # cột từ gốc
    COL_CHUAN       = "Dạng chuẩn hóa"    # cột dạng chuẩn hóa
    COL_CAM_XUC     = "Nhãn cảm xúc"       # nhãn cảm xúc của từ
    COL_DO_MANH     = "Độ mạnh (1=Nhẹ, 2=TB, 3=Mạnh)" # cường độ (vd: 1–5 hoặc low/mid/high)

    #Trọng số áp dụng khi cảm xúc của teencode khớp sentiment câu
    # Được dùng trong feature engineering (xem normalize_teencode_weighted)
    EMOTION_WEIGHT  = {
        "positive": 1.0,
        "negative": 1.0,
        "neutral":  0.5,
    }

cfg = Config()
torch.manual_seed(cfg.SEED)
np.random.seed(cfg.SEED)
print(f"Device: {cfg.DEVICE}")

LABEL2ID   = {"negative": 0, "neutral": 1, "positive": 2}
ID2LABEL   = {v: k for k, v in LABEL2ID.items()}
CLASS_NAMES = ["negative", "neutral", "positive"]


# 1. Load teencode dictionary  ← THAY ĐỔI CHÍNH
#    File mới có 4 cột: teencode | định dạng chuẩn | cảm xúc | độ mạnh
teencode_df = pd.read_excel(cfg.TEENCODE_PATH)

# Chuẩn hoá tên cột (bỏ khoảng trắng thừa ở đầu/cuối)
teencode_df.columns = teencode_df.columns.str.strip()

teencode_df = teencode_df.dropna(
    subset=[cfg.COL_TEENCODE, cfg.COL_CHUAN]
)

# Kiểm tra cột bắt buộc
required_cols = [cfg.COL_TEENCODE, cfg.COL_CHUAN, cfg.COL_CAM_XUC, cfg.COL_DO_MANH]
missing = [c for c in required_cols if c not in teencode_df.columns]
if missing:
    raise ValueError(
        f"File teencode thiếu cột: {missing}\n"
        f"Cột hiện có: {list(teencode_df.columns)}"
    )

#Từ điển chính: teencode → dạng chuẩn
TEENCODE_DICT: dict[str, str] = {}

#Từ điển mở rộng: teencode → {chuan, cam_xuc, do_manh}
TEENCODE_META: dict[str, dict] = {}

for _, row in teencode_df.iterrows():
    key  = str(row[cfg.COL_TEENCODE]).strip().lower()
    val  = str(row[cfg.COL_CHUAN]).strip().split("/")[0].strip()
    emotion = str(row[cfg.COL_CAM_XUC]).strip().lower() \
              if pd.notna(row[cfg.COL_CAM_XUC]) else "neutral"
    strength_raw = row[cfg.COL_DO_MANH]

    # Chuẩn hoá độ mạnh → float [0, 1]
    if pd.isna(strength_raw):
        strength = 0.5
    else:
        try:
            strength = float(strength_raw)
            # Nếu file lưu thang 1–5, chuẩn hoá về [0,1]
            if strength > 1.0:
                strength = (strength - 1) / 4.0
        except (ValueError, TypeError):
            # Có thể là string "low"/"mid"/"high"
            strength_map = {"low": 0.25, "mid": 0.5, "medium": 0.5,
                            "high": 0.75, "very high": 1.0, "thấp": 0.25,
                            "trung bình": 0.5, "cao": 0.75, "rất cao": 1.0}
            strength = strength_map.get(str(strength_raw).strip().lower(), 0.5)

    if key and val:
        TEENCODE_DICT[key] = val
        TEENCODE_META[key] = {
            "chuan":    val,
            "cam_xuc":  emotion,
            "do_manh":  strength,
        }

# Ưu tiên cụm nhiều từ (phrase) khi thay thế
PHRASE_LIST = sorted(
    [(k, v) for k, v in TEENCODE_DICT.items() if " " in k],
    key=lambda x: len(x[0]), reverse=True
)

print(f"Loaded {len(TEENCODE_DICT)} teencode entries.")
print(f"  • Có thông tin cảm xúc : {sum(1 for m in TEENCODE_META.values() if m['cam_xuc'] != 'neutral')} entries")
print(f"  • Cụm nhiều từ (phrase): {len(PHRASE_LIST)} entries")

# Thống kê phân bố cảm xúc trong từ điển teencode
emotion_dist = Counter(m["cam_xuc"] for m in TEENCODE_META.values())
print(f"  • Phân bố cảm xúc trong từ điển: {dict(emotion_dist)}")


# 2. Text preprocessing
def normalize_teencode(text: str) -> str:
    """
    Chuẩn hoá văn bản:
      1. Viết thường
      2. Rút ngắn ký tự lặp ≥ 3 → giữ 2
      3. Thay cụm nhiều từ (phrase) trước
      4. Thay từng token theo từ điển teencode
      5. Xoá khoảng trắng thừa
    """
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r"(.)\1{2,}", r"\1\1", text)

    # Thay cụm từ trước (ưu tiên khớp dài hơn)
    for phrase, replacement in PHRASE_LIST:
        text = text.replace(phrase, f" {replacement} ")

    # Thay từng token
    tokens = text.split()
    tokens = [TEENCODE_DICT.get(t, t) for t in tokens]
    text = " ".join(tokens)
    text = re.sub(r"\s+", " ", text).strip()
    return text


def extract_emotion_features(text: str) -> dict:
    """
    Trích xuất đặc trưng cảm xúc từ văn bản dựa trên metadata teencode.
    Trả về dict chứa:
      - emotion_score_pos  : tổng (độ_mạnh) của token positive
      - emotion_score_neg  : tổng (độ_mạnh) của token negative
      - emotion_score_neu  : tổng (độ_mạnh) của token neutral
      - n_emotion_tokens   : số token có trong TEENCODE_META
      - dominant_emotion   : nhãn cảm xúc chiếm ưu thế (string)

    Lưu ý: hàm này tính trên văn bản GỐC (trước normalize) để giữ
    chính xác token teencode gốc.
    """
    if not isinstance(text, str):
        return {
            "emotion_score_pos": 0.0, "emotion_score_neg": 0.0,
            "emotion_score_neu": 0.0, "n_emotion_tokens":  0,
            "dominant_emotion": "neutral",
        }
    tokens = text.lower().split()
    scores = {"positive": 0.0, "negative": 0.0, "neutral": 0.0}
    count  = 0

    for tok in tokens:
        meta = TEENCODE_META.get(tok)
        if meta:
            emo = meta["cam_xuc"]
            if emo not in scores:
                emo = "neutral"
            scores[emo] += meta["do_manh"]
            count += 1

    dominant = max(scores, key=scores.get)
    return {
        "emotion_score_pos": scores["positive"],
        "emotion_score_neg": scores["negative"],
        "emotion_score_neu": scores["neutral"],
        "n_emotion_tokens":  count,
        "dominant_emotion":  dominant,
    }


# 3. Load & preprocess data
train_df = pd.read_csv(os.path.join(cfg.DATA_DIR, "train.csv")).dropna()
valid_df  = pd.read_csv(os.path.join(cfg.DATA_DIR, "valid.csv")).dropna()
test_df   = pd.read_csv(os.path.join(cfg.DATA_DIR, "test.csv")).dropna()

print("Đang tiền xử lý teencode...")
train_texts = [normalize_teencode(s) for s in train_df["sentence"]]
valid_texts  = [normalize_teencode(s) for s in valid_df["sentence"]]
test_texts   = [normalize_teencode(s) for s in test_df["sentence"]]

train_labels = train_df["sentiment"].map(LABEL2ID).values
valid_labels  = valid_df["sentiment"].map(LABEL2ID).values
test_labels   = test_df["sentiment"].map(LABEL2ID).values

#Trích xuất đặc trưng cảm xúc từ văn bản gốc 
print("Đang trích xuất emotion features từ teencode metadata...")
train_emo = [extract_emotion_features(s) for s in train_df["sentence"]]
valid_emo  = [extract_emotion_features(s) for s in valid_df["sentence"]]
test_emo   = [extract_emotion_features(s) for s in test_df["sentence"]]

# Chuyển sang numpy array (4 chiều số): [pos_score, neg_score, neu_score, n_tokens]
def emo_to_array(emo_list):
    return np.array(
        [
            [
                e["emotion_score_pos"],
                e["emotion_score_neg"],
                e["emotion_score_neu"],
                e["n_emotion_tokens"],
            ]
            for e in emo_list
        ],
        dtype=np.float32,
    )

train_emo_arr = emo_to_array(train_emo)
valid_emo_arr  = emo_to_array(valid_emo)
test_emo_arr   = emo_to_array(test_emo)

print(f"Train: {len(train_texts)} | Valid: {len(valid_texts)} | Test: {len(test_texts)}")
print(f"Label dist train: {np.bincount(train_labels)}")

# Thống kê dominant emotion trong tập train
dom_train = Counter(e["dominant_emotion"] for e in train_emo)
print(f"Dominant emotion distribution (train): {dict(dom_train)}")


# 4. Build vocabulary từ tập train
PAD_TOKEN = "<PAD>"   # index 0
UNK_TOKEN = "<UNK>"   # index 1

def build_vocab(texts, max_vocab=cfg.VOCAB_SIZE):
    counter = Counter()
    for text in texts:
        counter.update(text.split())
    vocab = [PAD_TOKEN, UNK_TOKEN] + [w for w, _ in counter.most_common(max_vocab - 2)]
    word2idx = {w: i for i, w in enumerate(vocab)}
    return word2idx

word2idx = build_vocab(train_texts)
VOCAB_SIZE_ACTUAL = len(word2idx)
print(f"Vocab size: {VOCAB_SIZE_ACTUAL}")


def encode(text, word2idx, max_len):
    """Chuyển text → list id, padding/truncate về max_len."""
    tokens = text.split()[:max_len]
    ids = [word2idx.get(t, 1) for t in tokens]   # 1 = UNK
    ids += [0] * (max_len - len(ids))             # Pad bên phải
    return ids


# 5. Dataset & DataLoader
#    ← THAY ĐỔI: Dataset nhận thêm emotion_features (4 chiều)
class TextDataset(Dataset):
    def __init__(self, texts, labels, word2idx, max_len, emo_arr=None):
        self.encoded  = [encode(t, word2idx, max_len) for t in texts]
        self.labels   = labels
        # emo_arr: (N, 4) float32 — None nếu không dùng
        self.emo_arr  = emo_arr

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        x = torch.tensor(self.encoded[idx], dtype=torch.long)
        y = torch.tensor(self.labels[idx],  dtype=torch.long)
        if self.emo_arr is not None:
            emo = torch.tensor(self.emo_arr[idx], dtype=torch.float)
            return x, emo, y
        return x, y

train_dataset = TextDataset(train_texts, train_labels, word2idx,
                             cfg.MAX_LENGTH, train_emo_arr)
valid_dataset  = TextDataset(valid_texts,  valid_labels,  word2idx,
                             cfg.MAX_LENGTH, valid_emo_arr)
test_dataset   = TextDataset(test_texts,   test_labels,   word2idx,
                             cfg.MAX_LENGTH, test_emo_arr)

train_loader = DataLoader(train_dataset, batch_size=cfg.BATCH_SIZE, shuffle=True)
valid_loader  = DataLoader(valid_dataset,  batch_size=cfg.BATCH_SIZE)
test_loader   = DataLoader(test_dataset,   batch_size=cfg.BATCH_SIZE)


# 6. CNN Text Model
#    ← THAY ĐỔI: thêm nhánh Emotion Feature Fusion
#
#    Kiến trúc:
#      Embedding → Multi-scale Conv1d → MaxPool → Concat  ─┐
#                                                           ├→ Concat → Dropout → FC
#      EmotionFeatures (4-dim) → Linear(4→32) → ReLU      ─┘
EMO_DIM = 4   # [pos_score, neg_score, neu_score, n_tokens]

class TextCNN(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_filters,
                 kernel_sizes, num_classes, dropout,
                 emo_dim=EMO_DIM, emo_hidden=32, pad_idx=0):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim,
                                      padding_idx=pad_idx)

        # Mỗi kernel_size → 1 Conv layer (bắt n-gram features)
        self.convs = nn.ModuleList([
            nn.Conv1d(
                in_channels  = embed_dim,
                out_channels = num_filters,
                kernel_size  = k,
                padding      = k // 2,    # same padding
            )
            for k in kernel_sizes
        ])

        self.dropout = nn.Dropout(dropout)

        # Nhánh emotion feature từ teencode metadata
        self.emo_fc = nn.Linear(emo_dim, emo_hidden)

        # Fully-connected cuối: (num_filters * len(kernels)) + emo_hidden → num_classes
        cnn_out_dim = num_filters * len(kernel_sizes)
        self.fc = nn.Linear(cnn_out_dim + emo_hidden, num_classes)

        self._init_weights()

    def _init_weights(self):
        for conv in self.convs:
            nn.init.kaiming_normal_(conv.weight, nonlinearity="relu")
            nn.init.zeros_(conv.bias)
        nn.init.kaiming_normal_(self.emo_fc.weight, nonlinearity="relu")
        nn.init.zeros_(self.emo_fc.bias)
        nn.init.xavier_uniform_(self.fc.weight)
        nn.init.zeros_(self.fc.bias)

    def forward(self, x, emo):
        # x   : (batch, seq_len)
        # emo : (batch, emo_dim)

        #CNN branch
        emb = self.embedding(x)            # (batch, seq_len, embed_dim)
        emb = emb.permute(0, 2, 1)        # (batch, embed_dim, seq_len)

        pooled = []
        for conv in self.convs:
            c = F.relu(conv(emb))          # (batch, num_filters, seq_len')
            c = F.max_pool1d(c, c.size(2)) # (batch, num_filters, 1)
            c = c.squeeze(2)               # (batch, num_filters)
            pooled.append(c)

        cnn_out = torch.cat(pooled, dim=1) # (batch, num_filters * len(kernels))

        #Emotion feature branch 
        emo_out = F.relu(self.emo_fc(emo)) # (batch, emo_hidden)

        #Fusion & classification
        fused = torch.cat([cnn_out, emo_out], dim=1)  # concat hai nhánh
        fused = self.dropout(fused)
        out   = self.fc(fused)             # (batch, num_classes)
        return out


model = TextCNN(
    vocab_size   = VOCAB_SIZE_ACTUAL,
    embed_dim    = cfg.EMBED_DIM,
    num_filters  = cfg.NUM_FILTERS,
    kernel_sizes = cfg.KERNEL_SIZES,
    num_classes  = cfg.NUM_CLASSES,
    dropout      = cfg.DROPOUT,
    emo_dim      = EMO_DIM,
    emo_hidden   = 32,
).to(cfg.DEVICE)

# Class-weighted loss để xử lý mất cân bằng nhãn
class_counts  = np.bincount(train_labels)
class_weights = torch.tensor(
    1.0 / class_counts, dtype=torch.float
).to(cfg.DEVICE)
class_weights = class_weights / class_weights.sum() * cfg.NUM_CLASSES

criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = Adam(model.parameters(), lr=cfg.LR, weight_decay=1e-4)
scheduler = ReduceLROnPlateau(optimizer, mode="max", factor=0.5, patience=2)

total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\nModel parameters: {total_params:,}")
print(model)


# 7. Train / Eval helpers  ← THAY ĐỔI: forward nhận thêm emo tensor
def train_epoch(model, loader, criterion, optimizer):
    model.train()
    total_loss, total_correct, total = 0, 0, 0
    for x, emo, y in loader:
        x, emo, y = x.to(cfg.DEVICE), emo.to(cfg.DEVICE), y.to(cfg.DEVICE)
        optimizer.zero_grad()
        logits = model(x, emo)
        loss   = criterion(logits, y)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        total_loss    += loss.item() * x.size(0)
        total_correct += (logits.argmax(1) == y).sum().item()
        total         += x.size(0)

    return total_loss / total, total_correct / total


def eval_epoch(model, loader, criterion):
    model.eval()
    total_loss, total_correct, total = 0, 0, 0
    all_preds, all_labels = [], []

    with torch.no_grad():
        for x, emo, y in loader:
            x, emo, y = x.to(cfg.DEVICE), emo.to(cfg.DEVICE), y.to(cfg.DEVICE)
            logits = model(x, emo)
            loss   = criterion(logits, y)

            total_loss    += loss.item() * x.size(0)
            total_correct += (logits.argmax(1) == y).sum().item()
            total         += x.size(0)
            all_preds.extend(logits.argmax(1).cpu().numpy())
            all_labels.extend(y.cpu().numpy())

    f1 = f1_score(all_labels, all_preds, average="macro")
    return total_loss / total, total_correct / total, f1


def predict_proba(model, loader):
    """Trả về (probs, preds, labels)."""
    model.eval()
    all_probs, all_preds, all_labels = [], [], []
    with torch.no_grad():
        for x, emo, y in loader:
            x, emo = x.to(cfg.DEVICE), emo.to(cfg.DEVICE)
            logits = model(x, emo)
            probs  = torch.softmax(logits, dim=-1).cpu().numpy()
            preds  = logits.argmax(1).cpu().numpy()
            all_probs.extend(probs)
            all_preds.extend(preds)
            all_labels.extend(y.numpy())
    return np.array(all_probs), np.array(all_preds), np.array(all_labels)


# 8. Training loop với Early Stopping
history = {
    "train_loss": [], "train_acc": [],
    "val_loss":   [], "val_acc":   [], "val_f1": [],
}

best_val_f1  = -1
best_epoch   = 0
patience_cnt = 0
best_state   = None

print("\n" + "=" * 65)
print(f"{'Epoch':>5} {'Train Loss':>11} {'Train Acc':>10} "
      f"{'Val Loss':>10} {'Val Acc':>9} {'Val F1':>8}")
print("=" * 65)

for epoch in range(1, cfg.EPOCHS + 1):
    tr_loss, tr_acc           = train_epoch(model, train_loader, criterion, optimizer)
    val_loss, val_acc, val_f1 = eval_epoch(model, valid_loader, criterion)
    scheduler.step(val_f1)

    history["train_loss"].append(tr_loss)
    history["train_acc"].append(tr_acc)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)
    history["val_f1"].append(val_f1)

    marker = " ← best" if val_f1 > best_val_f1 else ""
    print(f"{epoch:>5} {tr_loss:>11.4f} {tr_acc:>10.4f} "
          f"{val_loss:>10.4f} {val_acc:>9.4f} {val_f1:>8.4f}{marker}")

    if val_f1 > best_val_f1:
        best_val_f1  = val_f1
        best_epoch   = epoch
        patience_cnt = 0
        best_state   = {k: v.cpu().clone() for k, v in model.state_dict().items()}
    else:
        patience_cnt += 1
        if patience_cnt >= cfg.PATIENCE:
            print(f"\nEarly stopping tại epoch {epoch} "
                  f"(best epoch={best_epoch}, best val F1={best_val_f1:.4f})")
            break

model.load_state_dict(best_state)
model.to(cfg.DEVICE)
print(f"\nĐã khôi phục best model từ epoch {best_epoch}.")


# 9. Plot training curves
epochs_ran = list(range(1, len(history["train_loss"]) + 1))

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

axes[0].plot(epochs_ran, history["train_loss"], label="Train Loss", marker="o")
axes[0].plot(epochs_ran, history["val_loss"],   label="Val Loss",   marker="s")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Loss")
axes[0].set_title("Loss Curves"); axes[0].legend(); axes[0].grid(True)

axes[1].plot(epochs_ran, history["train_acc"], label="Train Acc", marker="o")
axes[1].plot(epochs_ran, history["val_acc"],   label="Val Acc",   marker="s")
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Accuracy")
axes[1].set_title("Accuracy Curves"); axes[1].legend(); axes[1].grid(True)

axes[2].plot(epochs_ran, history["val_f1"], label="Val F1 (macro)",
             marker="^", color="green")
axes[2].axvline(best_epoch, linestyle="--", color="red",
                alpha=0.6, label=f"Best epoch={best_epoch}")
axes[2].set_xlabel("Epoch"); axes[2].set_ylabel("F1 Macro")
axes[2].set_title("Validation F1 Macro"); axes[2].legend(); axes[2].grid(True)

plt.suptitle("CNN Text + Teencode Emotion Features – Training History",
             fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("cnn_training_curves.png", dpi=150)
plt.show()


# 10. Evaluate on test set
y_pred_probs, y_pred, y_true = predict_proba(model, test_loader)

acc        = accuracy_score(y_true, y_pred)
prec_macro = precision_score(y_true, y_pred, average="macro")
rec_macro  = recall_score(y_true, y_pred, average="macro")
f1_macro   = f1_score(y_true, y_pred, average="macro")
y_true_bin = label_binarize(y_true, classes=[0, 1, 2])
auc_macro  = roc_auc_score(y_true_bin, y_pred_probs,
                            average="macro", multi_class="ovr")

print("\n" + "=" * 60)
print("FINAL TEST SET METRICS – CNN Text + Teencode Emotion Features")
print("=" * 60)
print(f"{'Metric':<25} {'Value':>10}")
print("-" * 60)
print(f"{'Accuracy':<25} {acc:>10.4f}")
print(f"{'Precision (macro)':<25} {prec_macro:>10.4f}")
print(f"{'Recall (macro)':<25} {rec_macro:>10.4f}")
print(f"{'F1-score (macro)':<25} {f1_macro:>10.4f}")
print(f"{'AUC (macro)':<25} {auc_macro:>10.4f}")
print("=" * 60)

print("\nCLASSIFICATION REPORT")
print(classification_report(y_true, y_pred, target_names=CLASS_NAMES))


# 11. Confusion Matrix
cf = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(7, 6))
sn.heatmap(cf, annot=True, fmt="d", cmap="Blues",
           xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
           cbar_kws={"label": "Count"})
plt.xlabel("Predicted", fontsize=12)
plt.ylabel("True", fontsize=12)
plt.title("Confusion Matrix – CNN Text (Test Set)", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("cnn_confusion_matrix.png", dpi=150)
plt.show()


# 12. Class-wise bar chart
prec_cls = precision_score(y_true, y_pred, average=None)
rec_cls  = recall_score(y_true, y_pred, average=None)
f1_cls   = f1_score(y_true, y_pred, average=None)

x = np.arange(len(CLASS_NAMES)); w = 0.25
plt.figure(figsize=(9, 5))
plt.bar(x - w, prec_cls, w, label="Precision", alpha=0.85)
plt.bar(x,     rec_cls,  w, label="Recall",    alpha=0.85)
plt.bar(x + w, f1_cls,   w, label="F1-score",  alpha=0.85)
plt.xticks(x, CLASS_NAMES)
plt.ylim([0, 1.1]); plt.ylabel("Score")
plt.title("Class-wise Metrics – CNN Text + Teencode Emotion Features",
          fontweight="bold")
plt.legend(); plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("cnn_classwise_metrics.png", dpi=150)
plt.show()


# 13. ROC Curves
fpr, tpr, roc_auc_d = {}, {}, {}
for i in range(3):
    fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_pred_probs[:, i])
    roc_auc_d[i]      = auc(fpr[i], tpr[i])
fpr["macro"], tpr["macro"], _ = roc_curve(y_true_bin.ravel(), y_pred_probs.ravel())
roc_auc_d["macro"]            = auc(fpr["macro"], tpr["macro"])

colors = ["#FF6B6B", "#4ECDC4", "#45B7D1"]
plt.figure(figsize=(9, 7))
for i in range(3):
    plt.plot(fpr[i], tpr[i], color=colors[i], lw=2,
             label=f"{CLASS_NAMES[i]} (AUC={roc_auc_d[i]:.4f})")
plt.plot(fpr["macro"], tpr["macro"],
         color="deeppink", linestyle="--", lw=2.5,
         label=f"macro-avg (AUC={roc_auc_d['macro']:.4f})")
plt.plot([0, 1], [0, 1], "k--", lw=1.5, alpha=0.5, label="Random")
plt.xlim([0, 1]); plt.ylim([0, 1.05])
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curves – CNN Text + Teencode Emotion Features", fontweight="bold")
plt.legend(loc="lower right"); plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("cnn_roc_curves.png", dpi=150)
plt.show()


# 14. Example predictions
print("\nEXAMPLE PREDICTIONS (first 10 test samples)")
print("-" * 75)
for i in range(min(10, len(test_texts))):
    orig      = test_df["sentence"].iloc[i]
    norm      = test_texts[i]
    emo_info  = test_emo[i]
    true_lbl  = ID2LABEL[y_true[i]]
    pred_lbl  = ID2LABEL[y_pred[i]]
    probs     = y_pred_probs[i]
    mark      = "✓" if true_lbl == pred_lbl else "✗"

    print(f"[{mark}] Gốc  : {orig[:60]}")
    if orig.lower() != norm:
        print(f"     Chuẩn: {norm[:60]}")
    print(f"     Emotion tokens: {emo_info['n_emotion_tokens']} "
          f"| dominant: {emo_info['dominant_emotion']} "
          f"| pos={emo_info['emotion_score_pos']:.2f} "
          f"neg={emo_info['emotion_score_neg']:.2f}")
    print(f"     True: {true_lbl:>8} | Pred: {pred_lbl:>8} | "
          f"neg={probs[0]:.3f} neu={probs[1]:.3f} pos={probs[2]:.3f}")
    print()


# 15. Lưu kết quả để so sánh
all_results = {}
all_results["CNN_Text_Emotion"] = {
    "accuracy":        acc,
    "precision_macro": prec_macro,
    "recall_macro":    rec_macro,
    "f1_macro":        f1_macro,
    "auc_macro":       auc_macro,
}
print("Stored in all_results:", all_results)
```

## ViSoBERT

```
"""
ViSoBERT – Phân tích cảm xúc tiếng Việt (3 lớp: negative / neutral / positive)
"""

# 0. Cài đặt thư viện (chạy đầu tiên trong Colab)
import subprocess, sys

def install(pkg):
    subprocess.check_call([sys.executable, "-m", "pip", "install", pkg, "-q"])

install("transformers[torch]")
install("accelerate")
install("sentencepiece")
install("openpyxl")  # để đọc .xlsx

# Imports
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
from collections import Counter

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from torch.optim.lr_scheduler import ReduceLROnPlateau, LambdaLR  # ← fix: bỏ SequentialLR

from transformers import AutoTokenizer, AutoModel

from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    confusion_matrix, roc_auc_score, roc_curve, auc,
    classification_report,
)
from sklearn.preprocessing import label_binarize

# 1. Config
class Config:
    TEENCODE_PATH   = "/content/drive/MyDrive/Do_an_co_so/DACS/data/teencode_final .xlsx"
    DATA_DIR        = "/content/drive/MyDrive/Do_an_co_so/DACS/data/dataset"

    SEED            = 42
    NUM_CLASSES     = 3
    MAX_LENGTH      = 128

    MODEL_NAME      = "uitnlp/visobert"

    BATCH_SIZE      = 16
    EPOCHS          = 10
    LR_BERT         = 2e-5
    LR_HEAD         = 1e-4
    WARMUP_RATIO    = 0.1
    WEIGHT_DECAY    = 1e-2
    DROPOUT         = 0.1
    PATIENCE        = 3

    DEVICE          = "cuda" if torch.cuda.is_available() else "cpu"

    COL_TEENCODE    = "Teencode"
    COL_NORMALIZED  = "Định dạng chuẩn"
    COL_EMOTION     = "Nhãn cảm xúc"
    COL_INTENSITY   = "Độ mạnh (1=Nhẹ, 2=TB, 3=Mạnh)"

    HIGH_INTENSITY  = {"cao", "rất cao", "high", "very high", "3", "4", "5"}
    REPEAT_STRONG   = 1

cfg = Config()
torch.manual_seed(cfg.SEED)
np.random.seed(cfg.SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(cfg.SEED)

print(f"Device : {cfg.DEVICE}")
print(f"Model  : {cfg.MODEL_NAME}")

LABEL2ID    = {"negative": 0, "neutral": 1, "positive": 2}
ID2LABEL    = {v: k for k, v in LABEL2ID.items()}
CLASS_NAMES = ["negative", "neutral", "positive"]

# 2. Load teencode dictionary
teencode_df = pd.read_excel(cfg.TEENCODE_PATH)

def _find_col(df, candidates):
    norm = {c.strip().lower(): c for c in df.columns}
    for cand in candidates:
        key = cand.strip().lower()
        if key in norm:
            return norm[key]
    return None

col_tc  = _find_col(teencode_df, [cfg.COL_TEENCODE,   "teencode"])
col_std = _find_col(teencode_df, [cfg.COL_NORMALIZED,  "dạng chuẩn hóa", "dinh dang chuan"])
col_emo = _find_col(teencode_df, [cfg.COL_EMOTION,     "cam xuc", "emotion"])
col_int = _find_col(teencode_df, [cfg.COL_INTENSITY,   "do manh", "intensity", "mức độ"])

missing = [n for n, c in [(cfg.COL_TEENCODE, col_tc), (cfg.COL_NORMALIZED, col_std),
                           (cfg.COL_EMOTION, col_emo), (cfg.COL_INTENSITY, col_int)] if c is None]
if missing:
    print(f"[WARN] Không tìm thấy cột: {missing}")
    print(f"       Các cột hiện có   : {list(teencode_df.columns)}")

required_cols = [c for c in [col_tc, col_std] if c is not None]
teencode_df   = teencode_df.dropna(subset=required_cols)

TEENCODE_DICT      = {}
TEENCODE_EMOTION   = {}
TEENCODE_INTENSITY = {}

for _, row in teencode_df.iterrows():
    key = str(row[col_tc]).strip().lower()
    if not key:
        continue
    val = str(row[col_std]).strip().split("/")[0].strip() if col_std else key
    if not val:
        continue
    TEENCODE_DICT[key] = val
    if col_emo and pd.notna(row[col_emo]):
        TEENCODE_EMOTION[key] = str(row[col_emo]).strip().lower()
    if col_int and pd.notna(row[col_int]):
        TEENCODE_INTENSITY[key] = str(row[col_int]).strip().lower()

PHRASE_LIST = sorted(
    [(k, v) for k, v in TEENCODE_DICT.items() if " " in k],
    key=lambda x: len(x[0]), reverse=True
)

print(f"Loaded {len(TEENCODE_DICT)} teencode entries.")
print(f"  → Cảm xúc  : {len(TEENCODE_EMOTION)} | Độ mạnh: {len(TEENCODE_INTENSITY)}")
if TEENCODE_EMOTION:
    print(f"  → Phân bố  : {dict(Counter(TEENCODE_EMOTION.values()).most_common(8))}")

# 3. Text preprocessing
def _is_high_intensity(key: str) -> bool:
    return TEENCODE_INTENSITY.get(key, "") in cfg.HIGH_INTENSITY


def normalize_teencode(text: str) -> str:
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r"(.)\1{2,}", r"\1\1", text)
    for phrase, replacement in PHRASE_LIST:
        text = text.replace(phrase, f" {replacement} ")
    tokens = text.split()
    new_tokens = []
    for t in tokens:
        replaced = TEENCODE_DICT.get(t, t)
        new_tokens.append(replaced)
        if _is_high_intensity(t):
            new_tokens.extend([replaced] * cfg.REPEAT_STRONG)
    return re.sub(r"\s+", " ", " ".join(new_tokens)).strip()

# 4. Load & preprocess data
train_df = pd.read_csv(os.path.join(cfg.DATA_DIR, "train.csv")).dropna()
valid_df  = pd.read_csv(os.path.join(cfg.DATA_DIR, "valid.csv")).dropna()
test_df   = pd.read_csv(os.path.join(cfg.DATA_DIR, "test.csv")).dropna()

print("Đang tiền xử lý teencode...")
train_texts = [normalize_teencode(s) for s in train_df["sentence"]]
valid_texts  = [normalize_teencode(s) for s in valid_df["sentence"]]
test_texts   = [normalize_teencode(s) for s in test_df["sentence"]]

train_labels = train_df["sentiment"].map(LABEL2ID).values.astype(int)
valid_labels  = valid_df["sentiment"].map(LABEL2ID).values.astype(int)
test_labels   = test_df["sentiment"].map(LABEL2ID).values.astype(int)

print(f"Train: {len(train_texts)} | Valid: {len(valid_texts)} | Test: {len(test_texts)}")
print(f"Label dist train : {np.bincount(train_labels)}")
print(f"Label dist valid : {np.bincount(valid_labels)}")
print(f"Label dist test  : {np.bincount(test_labels)}")

# 5. Tokenizer
print(f"\nLoading tokenizer: {cfg.MODEL_NAME} ...")
tokenizer = AutoTokenizer.from_pretrained(cfg.MODEL_NAME)

# 6. Dataset & DataLoader
class SentimentDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len):
        self.texts     = texts
        self.labels    = labels
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        enc = self.tokenizer(
            self.texts[idx],
            max_length      = self.max_len,
            padding         = "max_length",
            truncation      = True,
            return_tensors  = "pt",
        )
        return {
            "input_ids":      enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "label":          torch.tensor(self.labels[idx], dtype=torch.long),
        }

train_dataset = SentimentDataset(train_texts, train_labels, tokenizer, cfg.MAX_LENGTH)
valid_dataset  = SentimentDataset(valid_texts,  valid_labels,  tokenizer, cfg.MAX_LENGTH)
test_dataset   = SentimentDataset(test_texts,   test_labels,   tokenizer, cfg.MAX_LENGTH)

train_loader = DataLoader(train_dataset, batch_size=cfg.BATCH_SIZE, shuffle=True,  drop_last=False)
valid_loader  = DataLoader(valid_dataset,  batch_size=cfg.BATCH_SIZE)
test_loader   = DataLoader(test_dataset,   batch_size=cfg.BATCH_SIZE)

# 7. Model
class ViSoBERTClassifier(nn.Module):
    def __init__(self, model_name, num_classes, dropout):
        super().__init__()
        self.bert = AutoModel.from_pretrained(model_name)
        hidden    = self.bert.config.hidden_size  # 768

        self.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(hidden, hidden // 3),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden // 3, num_classes),
        )
        self._init_classifier()

    def _init_classifier(self):
        for layer in self.classifier:
            if isinstance(layer, nn.Linear):
                nn.init.xavier_uniform_(layer.weight)
                nn.init.zeros_(layer.bias)

    def forward(self, input_ids, attention_mask):
        outputs   = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        cls_token = outputs.last_hidden_state[:, 0, :]
        logits    = self.classifier(cls_token)
        attn_last = outputs.attentions
        return logits, attn_last

model = ViSoBERTClassifier(
    model_name  = cfg.MODEL_NAME,
    num_classes = cfg.NUM_CLASSES,
    dropout     = cfg.DROPOUT,
).to(cfg.DEVICE)

bert_params = list(model.bert.parameters())
head_params = list(model.classifier.parameters())

optimizer = AdamW([
    {"params": bert_params, "lr": cfg.LR_BERT, "weight_decay": cfg.WEIGHT_DECAY},
    {"params": head_params, "lr": cfg.LR_HEAD, "weight_decay": 0.0},
])

class_counts  = np.bincount(train_labels)
class_weights = torch.tensor(1.0 / class_counts, dtype=torch.float)
class_weights = (class_weights / class_weights.sum() * cfg.NUM_CLASSES).to(cfg.DEVICE)
criterion     = nn.CrossEntropyLoss(weight=class_weights)

total_steps  = len(train_loader) * cfg.EPOCHS
warmup_steps = int(total_steps * cfg.WARMUP_RATIO)

def lr_lambda(current_step):
    if current_step < warmup_steps:
        return float(current_step) / float(max(1, warmup_steps))
    return 1.0

warmup_scheduler = LambdaLR(optimizer, lr_lambda)
reduce_scheduler = ReduceLROnPlateau(optimizer, mode="max", factor=0.5, patience=2)

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\nTrainable parameters: {trainable:,}")

# 8. Train / Eval helpers
def train_epoch(model, loader, criterion, optimizer, warmup_sched, global_step, warmup_steps):
    model.train()
    total_loss = total_correct = total = 0

    for batch in loader:
        input_ids      = batch["input_ids"].to(cfg.DEVICE)
        attention_mask = batch["attention_mask"].to(cfg.DEVICE)
        labels         = batch["label"].to(cfg.DEVICE)

        optimizer.zero_grad()
        logits, _ = model(input_ids, attention_mask)
        loss      = criterion(logits, labels)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

        global_step += 1
        if global_step <= warmup_steps:
            warmup_sched.step()

        total_loss    += loss.item() * input_ids.size(0)
        total_correct += (logits.argmax(1) == labels).sum().item()
        total         += input_ids.size(0)

    return total_loss / total, total_correct / total, global_step


def eval_epoch(model, loader, criterion):
    model.eval()
    total_loss = total_correct = total = 0
    all_preds, all_labels = [], []

    with torch.no_grad():
        for batch in loader:
            input_ids      = batch["input_ids"].to(cfg.DEVICE)
            attention_mask = batch["attention_mask"].to(cfg.DEVICE)
            labels         = batch["label"].to(cfg.DEVICE)

            logits, _ = model(input_ids, attention_mask)
            loss      = criterion(logits, labels)

            total_loss    += loss.item() * input_ids.size(0)
            total_correct += (logits.argmax(1) == labels).sum().item()
            total         += input_ids.size(0)
            all_preds.extend(logits.argmax(1).cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    f1 = f1_score(all_labels, all_preds, average="macro")
    return total_loss / total, total_correct / total, f1


def predict_all(model, loader):
    model.eval()
    all_probs, all_preds, all_labels = [], [], []

    with torch.no_grad():
        for batch in loader:
            input_ids      = batch["input_ids"].to(cfg.DEVICE)
            attention_mask = batch["attention_mask"].to(cfg.DEVICE)
            labels         = batch["label"]

            logits, _ = model(input_ids, attention_mask)
            probs     = torch.softmax(logits, dim=-1).cpu().numpy()
            preds     = logits.argmax(1).cpu().numpy()

            all_probs.extend(probs)
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())

    return np.array(all_probs), np.array(all_preds), np.array(all_labels)

# 9. Training loop với Early Stopping
history = {
    "train_loss": [], "train_acc": [],
    "val_loss":   [], "val_acc":   [], "val_f1": [],
}

best_val_f1  = -1
best_epoch   = 0
patience_cnt = 0
best_state   = None
global_step  = 0  # ← fix: khởi tạo đúng chỗ, không dùng list

print("\n" + "=" * 68)
print(f"{'Ep':>4} {'TrLoss':>9} {'TrAcc':>8} "
      f"{'ValLoss':>9} {'ValAcc':>8} {'ValF1':>8} {'LR_bert':>12}")
print("=" * 68)

for epoch in range(1, cfg.EPOCHS + 1):
    tr_loss, tr_acc, global_step = train_epoch(
        model, train_loader, criterion, optimizer,
        warmup_scheduler, global_step, warmup_steps
    )
    val_loss, val_acc, val_f1 = eval_epoch(model, valid_loader, criterion)

    if global_step > warmup_steps:
        reduce_scheduler.step(val_f1)

    current_lr = optimizer.param_groups[0]["lr"]

    history["train_loss"].append(tr_loss)
    history["train_acc"].append(tr_acc)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)
    history["val_f1"].append(val_f1)

    marker = " ★" if val_f1 > best_val_f1 else ""
    print(f"{epoch:>4} {tr_loss:>9.4f} {tr_acc:>8.4f} "
          f"{val_loss:>9.4f} {val_acc:>8.4f} {val_f1:>8.4f} "
          f"{current_lr:>12.2e}{marker}")

    if val_f1 > best_val_f1:
        best_val_f1  = val_f1
        best_epoch   = epoch
        patience_cnt = 0
        best_state   = {k: v.cpu().clone() for k, v in model.state_dict().items()}
    else:
        patience_cnt += 1
        if patience_cnt >= cfg.PATIENCE:
            print(f"\nEarly stopping tại epoch {epoch} "
                  f"(best epoch={best_epoch}, val F1={best_val_f1:.4f})")
            break

model.load_state_dict(best_state)
model.to(cfg.DEVICE)
print(f"\nĐã khôi phục best model từ epoch {best_epoch}.")

torch.save({
    "model_state": best_state,
    "config":      cfg,
    "best_epoch":  best_epoch,
    "best_val_f1": best_val_f1,
}, "visobert_best.pt")
print("Đã lưu model → visobert_best.pt")

# 10. Plot training curves
epochs_ran = list(range(1, len(history["train_loss"]) + 1))
fig, axes  = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(epochs_ran, history["train_loss"], marker="o", label="Train Loss")
axes[0].plot(epochs_ran, history["val_loss"],   marker="s", label="Val Loss")
axes[0].axvline(best_epoch, linestyle="--", color="red", alpha=0.5,
                label=f"Best epoch={best_epoch}")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Loss")
axes[0].set_title("Loss Curves"); axes[0].legend(); axes[0].grid(True)

axes[1].plot(epochs_ran, history["train_acc"], marker="o", label="Train Acc")
axes[1].plot(epochs_ran, history["val_acc"],   marker="s", label="Val Acc")
axes[1].axvline(best_epoch, linestyle="--", color="red", alpha=0.5)
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Accuracy")
axes[1].set_title("Accuracy Curves"); axes[1].legend(); axes[1].grid(True)

axes[2].plot(epochs_ran, history["val_f1"], marker="^",
             color="green", label="Val F1 macro")
axes[2].axvline(best_epoch, linestyle="--", color="red", alpha=0.5,
                label=f"Best epoch={best_epoch}")
axes[2].set_xlabel("Epoch"); axes[2].set_ylabel("F1 Macro")
axes[2].set_title("Validation F1 Macro"); axes[2].legend(); axes[2].grid(True)

plt.suptitle("ViSoBERT – Training History", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("visobert_training_curves.png", dpi=150)
plt.show()

# 11. Evaluate on test set
y_pred_probs, y_pred, y_true = predict_all(model, test_loader)

acc        = accuracy_score(y_true, y_pred)
prec_macro = precision_score(y_true, y_pred, average="macro")
rec_macro  = recall_score(y_true, y_pred, average="macro")
f1_macro   = f1_score(y_true, y_pred, average="macro")
y_true_bin = label_binarize(y_true, classes=[0, 1, 2])
auc_macro  = roc_auc_score(y_true_bin, y_pred_probs,
                            average="macro", multi_class="ovr")

print("\n" + "=" * 60)
print("FINAL TEST SET METRICS – ViSoBERT + Teencode (4 cột)")
print("=" * 60)
print(f"{'Metric':<25} {'Value':>10}")
print("-" * 60)
print(f"{'Accuracy':<25} {acc:>10.4f}")
print(f"{'Precision (macro)':<25} {prec_macro:>10.4f}")
print(f"{'Recall (macro)':<25} {rec_macro:>10.4f}")
print(f"{'F1-score (macro)':<25} {f1_macro:>10.4f}")
print(f"{'AUC (macro)':<25} {auc_macro:>10.4f}")
print("=" * 60)
print("\nCLASSIFICATION REPORT")
print(classification_report(y_true, y_pred, target_names=CLASS_NAMES))

# 12. Confusion Matrix
cf = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(7, 6))
sn.heatmap(cf, annot=True, fmt="d", cmap="Blues",
           xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
           cbar_kws={"label": "Count"})
plt.xlabel("Predicted", fontsize=12); plt.ylabel("True", fontsize=12)
plt.title("Confusion Matrix – ViSoBERT (Test Set)", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("visobert_confusion_matrix.png", dpi=150)
plt.show()

# 13. Class-wise bar chart
prec_cls = precision_score(y_true, y_pred, average=None)
rec_cls  = recall_score(y_true, y_pred, average=None)
f1_cls   = f1_score(y_true, y_pred, average=None)

x = np.arange(len(CLASS_NAMES)); w = 0.25
plt.figure(figsize=(9, 5))
plt.bar(x - w, prec_cls, w, label="Precision", alpha=0.85)
plt.bar(x,     rec_cls,  w, label="Recall",    alpha=0.85)
plt.bar(x + w, f1_cls,   w, label="F1-score",  alpha=0.85)
plt.xticks(x, CLASS_NAMES)
plt.ylim([0, 1.1]); plt.ylabel("Score")
plt.title("Class-wise Metrics – ViSoBERT + Teencode (4 cột)", fontweight="bold")
plt.legend(); plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("visobert_classwise_metrics.png", dpi=150)
plt.show()

# 14. ROC Curves
fpr, tpr, roc_auc_d = {}, {}, {}
for i in range(3):
    fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_pred_probs[:, i])
    roc_auc_d[i]      = auc(fpr[i], tpr[i])
fpr["macro"], tpr["macro"], _ = roc_curve(y_true_bin.ravel(), y_pred_probs.ravel())
roc_auc_d["macro"] = auc(fpr["macro"], tpr["macro"])

colors = ["#FF6B6B", "#4ECDC4", "#45B7D1"]
plt.figure(figsize=(9, 7))
for i in range(3):
    plt.plot(fpr[i], tpr[i], color=colors[i], lw=2,
             label=f"{CLASS_NAMES[i]} (AUC={roc_auc_d[i]:.4f})")
plt.plot(fpr["macro"], tpr["macro"], color="deeppink", linestyle="--", lw=2.5,
         label=f"macro-avg (AUC={roc_auc_d['macro']:.4f})")
plt.plot([0, 1], [0, 1], "k--", lw=1.5, alpha=0.5, label="Random")
plt.xlim([0, 1]); plt.ylim([0, 1.05])
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curves – ViSoBERT + Teencode (4 cột)", fontweight="bold")
plt.legend(loc="lower right"); plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("visobert_roc_curves.png", dpi=150)
plt.show()

# 15. BERT Attention Heatmap  — fix: thêm try/except cho attn_implementation
def plot_bert_attention(model, tokenizer, texts, y_true, y_pred,
                        max_len, n=5, layer=-1, head=0):
    model.eval()
    _orig_bert = model.bert

    #fix: thử eager trước, fallback nếu không hỗ trợ
    try:
        _attn_bert = AutoModel.from_pretrained(
            cfg.MODEL_NAME,
            attn_implementation="eager",
        ).to(cfg.DEVICE)
    except TypeError:
        # Phiên bản transformers cũ không có tham số attn_implementation
        _attn_bert = AutoModel.from_pretrained(cfg.MODEL_NAME).to(cfg.DEVICE)

    model.bert = _attn_bert
    model.bert.eval()

    fig, axes = plt.subplots(n, 1, figsize=(16, n * 3.5))
    if n == 1:
        axes = [axes]

    for i in range(n):
        enc = tokenizer(
            texts[i],
            max_length     = max_len,
            padding        = "max_length",
            truncation     = True,
            return_tensors = "pt",
        )
        input_ids      = enc["input_ids"].to(cfg.DEVICE)
        attention_mask = enc["attention_mask"].to(cfg.DEVICE)

        with torch.no_grad():
            outputs = model.bert(
                input_ids         = input_ids,
                attention_mask    = attention_mask,
                output_attentions = True,
            )

        # fix: guard nếu attentions là None
        if outputs.attentions is None:
            print(f"[WARN] Attention weights không khả dụng cho mẫu {i}.")
            continue

        attn   = outputs.attentions[layer][0, head].cpu().numpy()
        tokens = tokenizer.convert_ids_to_tokens(input_ids[0].cpu())
        seq_len = int(attention_mask[0].sum().item())
        tokens  = tokens[:seq_len]
        attn    = attn[:seq_len, :seq_len]

        true_lbl = ID2LABEL[y_true[i]]
        pred_lbl = ID2LABEL[y_pred[i]]
        mark     = "✓" if true_lbl == pred_lbl else "✗"

        ax = axes[i]
        im = ax.imshow(attn, cmap="Blues", aspect="auto")
        ax.set_xticks(range(len(tokens)))
        ax.set_xticklabels(tokens, rotation=45, ha="right", fontsize=7)
        ax.set_yticks(range(len(tokens)))
        ax.set_yticklabels(tokens, fontsize=7)
        ax.set_title(
            f"[{mark}] True: {true_lbl} | Pred: {pred_lbl}  —  {texts[i][:55]}",
            fontsize=9,
        )
        plt.colorbar(im, ax=ax, fraction=0.02)

    plt.suptitle(f"BERT Attention Heatmap – Layer {layer}, Head {head}",
                 fontsize=13, fontweight="bold")
    plt.tight_layout()
    plt.savefig("visobert_attention_heatmap.png", dpi=150)
    plt.show()

    model.bert = _orig_bert  # khôi phục


#  fix: gọi hàm attention (bị thiếu trong bản gốc)
plot_bert_attention(
    model, tokenizer, test_texts, y_true, y_pred,
    max_len=cfg.MAX_LENGTH, n=5, layer=-1, head=0
)

# 16. Example predictions
print("\nEXAMPLE PREDICTIONS (first 10 test samples)")
print("-" * 70)
for i in range(min(10, len(test_texts))):
    orig     = test_df["sentence"].iloc[i]
    norm     = test_texts[i]
    true_lbl = ID2LABEL[y_true[i]]
    pred_lbl = ID2LABEL[y_pred[i]]
    probs    = y_pred_probs[i]
    mark     = "✓" if true_lbl == pred_lbl else "✗"
    print(f"[{mark}] Gốc  : {orig[:60]}")
    if orig.lower() != norm:
        print(f"     Chuẩn: {norm[:60]}")
    print(f"     True: {true_lbl:>8} | Pred: {pred_lbl:>8} | "
          f"neg={probs[0]:.3f} neu={probs[1]:.3f} pos={probs[2]:.3f}")
    print()

all_results = {}
all_results["ViSoBERT"] = {
    "accuracy":        acc,
    "precision_macro": prec_macro,
    "recall_macro":    rec_macro,
    "f1_macro":        f1_macro,
    "auc_macro":       auc_macro,
}
print("Stored in all_results:", all_results)
```

## mDeBERTa

```
"""
mDeBERTa-v3-base – Phân tích cảm xúc tiếng Việt (3 lớp: negative / neutral / positive)
Model: microsoft/mdeberta-v3-base
Pipeline khớp hoàn toàn với bản BiLSTM + ViSoBERT:
  - Teencode 4 cột (Teencode, Định dạng chuẩn, Cảm xúc, Độ mạnh)
  - Class-weighted CrossEntropy
  - Differential LR + Linear Warmup + ReduceLROnPlateau
  - Early stopping theo Val F1 macro
  - Đầy đủ visualization: curves, confusion matrix, ROC, class-wise bar, attention heatmap
"""

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sn
from collections import Counter

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from torch.optim.lr_scheduler import LambdaLR, ReduceLROnPlateau

from transformers import AutoTokenizer, AutoModel

from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    confusion_matrix, roc_auc_score, roc_curve, auc,
    classification_report,
)
from sklearn.preprocessing import label_binarize

# 0. Config
class Config:
    TEENCODE_PATH   = "/content/drive/MyDrive/Do_an_co_so/DACS/data/teencode_final .xlsx"
    DATA_DIR        = "/content/drive/MyDrive/Do_an_co_so/DACS/data/dataset"

    SEED            = 42
    NUM_CLASSES     = 3

    # mDeBERTa-v3-base
    MODEL_NAME      = "microsoft/mdeberta-v3-base"
    MAX_LENGTH      = 128       # mDeBERTa hỗ trợ 512; 128 đủ cho câu cảm xúc ngắn

    # Fine-tuning
    BATCH_SIZE      = 16
    GRAD_ACCUM      = 2         # effective batch = 16 × 2 = 32
    EPOCHS          = 10
    LR_DEBERTA      = 2e-5      # BERT-style lr cho encoder
    LR_HEAD         = 1e-4      # classifier head
    WARMUP_RATIO    = 0.1
    WEIGHT_DECAY    = 1e-2
    DROPOUT         = 0.1
    PATIENCE        = 3

    DEVICE          = "cuda" if torch.cuda.is_available() else "cpu"

    # Teencode – tên cột
    COL_TEENCODE    = "Teencode"
    COL_NORMALIZED  = "Định dạng chuẩn"
    COL_EMOTION     = "Nhãn cảm xúc"
    COL_INTENSITY   = "Độ mạnh (1=Nhẹ, 2=TB, 3=Mạnh)"

    HIGH_INTENSITY  = {"cao", "rất cao", "high", "very high", "3", "4", "5"}
    REPEAT_STRONG   = 1

cfg = Config()
torch.manual_seed(cfg.SEED)
np.random.seed(cfg.SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(cfg.SEED)

print(f"Device : {cfg.DEVICE}")
print(f"Model  : {cfg.MODEL_NAME}")

LABEL2ID    = {"negative": 0, "neutral": 1, "positive": 2}
ID2LABEL    = {v: k for k, v in LABEL2ID.items()}
CLASS_NAMES = ["negative", "neutral", "positive"]

# 1. Load teencode dictionary — 4 cột
teencode_df = pd.read_excel(cfg.TEENCODE_PATH)

def _find_col(df, candidates):
    norm = {c.strip().lower(): c for c in df.columns}
    for cand in candidates:
        if cand.strip().lower() in norm:
            return norm[cand.strip().lower()]
    return None

col_tc  = _find_col(teencode_df, [cfg.COL_TEENCODE,   "teencode"])
col_std = _find_col(teencode_df, [cfg.COL_NORMALIZED,  "dạng chuẩn hóa", "dinh dang chuan"])
col_emo = _find_col(teencode_df, [cfg.COL_EMOTION,     "cam xuc", "emotion"])
col_int = _find_col(teencode_df, [cfg.COL_INTENSITY,   "do manh", "intensity", "mức độ"])

missing = [n for n, c in [(cfg.COL_TEENCODE, col_tc), (cfg.COL_NORMALIZED, col_std),
                           (cfg.COL_EMOTION, col_emo), (cfg.COL_INTENSITY, col_int)] if c is None]
if missing:
    print(f"[WARN] Không tìm thấy cột: {missing}")
    print(f"       Các cột hiện có   : {list(teencode_df.columns)}")

teencode_df = teencode_df.dropna(subset=[c for c in [col_tc, col_std] if c is not None])

TEENCODE_DICT      = {}
TEENCODE_EMOTION   = {}
TEENCODE_INTENSITY = {}

for _, row in teencode_df.iterrows():
    key = str(row[col_tc]).strip().lower()
    if not key:
        continue
    val = str(row[col_std]).strip().split("/")[0].strip() if col_std else key
    if not val:
        continue
    TEENCODE_DICT[key] = val
    if col_emo and pd.notna(row[col_emo]):
        TEENCODE_EMOTION[key] = str(row[col_emo]).strip().lower()
    if col_int and pd.notna(row[col_int]):
        TEENCODE_INTENSITY[key] = str(row[col_int]).strip().lower()

PHRASE_LIST = sorted(
    [(k, v) for k, v in TEENCODE_DICT.items() if " " in k],
    key=lambda x: len(x[0]), reverse=True
)

print(f"Loaded {len(TEENCODE_DICT)} teencode entries.")
print(f"  → Cảm xúc: {len(TEENCODE_EMOTION)} | Độ mạnh: {len(TEENCODE_INTENSITY)}")
if TEENCODE_EMOTION:
    print(f"  → Phân bố: {dict(Counter(TEENCODE_EMOTION.values()).most_common(8))}")

# 2. Text preprocessing
def _is_high_intensity(key: str) -> bool:
    return TEENCODE_INTENSITY.get(key, "") in cfg.HIGH_INTENSITY


def normalize_teencode(text: str) -> str:
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r"(.)\1{2,}", r"\1\1", text)
    for phrase, replacement in PHRASE_LIST:
        text = text.replace(phrase, f" {replacement} ")
    tokens = text.split()
    new_tokens = []
    for t in tokens:
        replaced = TEENCODE_DICT.get(t, t)
        new_tokens.append(replaced)
        if _is_high_intensity(t):
            new_tokens.extend([replaced] * cfg.REPEAT_STRONG)
    return re.sub(r"\s+", " ", " ".join(new_tokens)).strip()

# 3. Load & preprocess data
train_df = pd.read_csv(os.path.join(cfg.DATA_DIR, "train.csv")).dropna()
valid_df  = pd.read_csv(os.path.join(cfg.DATA_DIR, "valid.csv")).dropna()
test_df   = pd.read_csv(os.path.join(cfg.DATA_DIR, "test.csv")).dropna()

print("Đang tiền xử lý teencode...")
train_texts = [normalize_teencode(s) for s in train_df["sentence"]]
valid_texts  = [normalize_teencode(s) for s in valid_df["sentence"]]
test_texts   = [normalize_teencode(s) for s in test_df["sentence"]]

train_labels = train_df["sentiment"].map(LABEL2ID).values.astype(int)
valid_labels  = valid_df["sentiment"].map(LABEL2ID).values.astype(int)
test_labels   = test_df["sentiment"].map(LABEL2ID).values.astype(int)

print(f"Train: {len(train_texts)} | Valid: {len(valid_texts)} | Test: {len(test_texts)}")
print(f"Label dist train : {np.bincount(train_labels)}")
print(f"Label dist valid : {np.bincount(valid_labels)}")
print(f"Label dist test  : {np.bincount(test_labels)}")

# 4. Tokenizer
print(f"\nLoading tokenizer: {cfg.MODEL_NAME} ...")
tokenizer = AutoTokenizer.from_pretrained(cfg.MODEL_NAME)

# 5. Dataset & DataLoader
class SentimentDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len):
        self.texts     = texts
        self.labels    = labels
        self.tokenizer = tokenizer
        self.max_len   = max_len

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        enc = self.tokenizer(
            self.texts[idx],
            max_length     = self.max_len,
            padding        = "max_length",
            truncation     = True,
            return_tensors = "pt",
        )
        return {
            "input_ids":      enc["input_ids"].squeeze(0),
            "attention_mask": enc["attention_mask"].squeeze(0),
            "token_type_ids": enc.get("token_type_ids",
                                      torch.zeros(self.max_len, dtype=torch.long)).squeeze(0),
            "label":          torch.tensor(self.labels[idx], dtype=torch.long),
        }

train_dataset = SentimentDataset(train_texts, train_labels, tokenizer, cfg.MAX_LENGTH)
valid_dataset  = SentimentDataset(valid_texts,  valid_labels,  tokenizer, cfg.MAX_LENGTH)
test_dataset   = SentimentDataset(test_texts,   test_labels,   tokenizer, cfg.MAX_LENGTH)

train_loader = DataLoader(train_dataset, batch_size=cfg.BATCH_SIZE, shuffle=True,  drop_last=False)
valid_loader  = DataLoader(valid_dataset,  batch_size=cfg.BATCH_SIZE)
test_loader   = DataLoader(test_dataset,   batch_size=cfg.BATCH_SIZE)

# 6. Model
class mDeBERTaClassifier(nn.Module):
    """
    mDeBERTa-v3-base + Classifier Head

    Tại sao KHÔNG dùng [CLS] như ViSoBERT?
    ─────────────────────────────────────────
    mDeBERTa-v3 dùng kiến trúc Disentangled Attention (nội dung + vị trí riêng)
    kết hợp với Enhanced Mask Decoder (EMD). Token [CLS] trong mDeBERTa
    không nhất thiết mang full thông tin ngữ nghĩa câu tốt nhất.
    → Mean-pool toàn bộ token (bỏ padding) thường mạnh hơn cho classification.

    Kiến trúc:
      mean_pool(last_hidden_state)  →  LayerNorm
        →  Dropout  →  Linear(768→384)  →  GELU
        →  Dropout  →  Linear(384→3)
    """
    def __init__(self, model_name, num_classes, dropout):
        super().__init__()
        self.deberta = AutoModel.from_pretrained(model_name)
        hidden       = self.deberta.config.hidden_size  # 768

        self.norm       = nn.LayerNorm(hidden)
        self.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(hidden, hidden // 2),   # 768 → 384
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(hidden // 2, num_classes),
        )
        self._init_classifier()

    def _init_classifier(self):
        for layer in self.classifier:
            if isinstance(layer, nn.Linear):
                nn.init.xavier_uniform_(layer.weight)
                nn.init.zeros_(layer.bias)

    def mean_pool(self, last_hidden_state, attention_mask):
        """Mean pooling — bỏ qua padding tokens."""
        mask_expanded = attention_mask.unsqueeze(-1).float()          # (B, L, 1)
        sum_hidden    = (last_hidden_state * mask_expanded).sum(1)    # (B, H)
        count         = mask_expanded.sum(1).clamp(min=1e-9)          # (B, 1)
        return sum_hidden / count                                      # (B, H)

    def forward(self, input_ids, attention_mask, token_type_ids=None):
        kwargs = dict(input_ids=input_ids, attention_mask=attention_mask)
        # token_type_ids: mDeBERTa có thể không dùng, truyền nếu có
        if token_type_ids is not None and token_type_ids.any():
            kwargs["token_type_ids"] = token_type_ids

        outputs    = self.deberta(**kwargs)
        pooled     = self.mean_pool(outputs.last_hidden_state, attention_mask)
        pooled     = self.norm(pooled)
        logits     = self.classifier(pooled)
        return logits, outputs.attentions   # attentions = None unless output_attentions=True

model = mDeBERTaClassifier(
    model_name  = cfg.MODEL_NAME,
    num_classes = cfg.NUM_CLASSES,
    dropout     = cfg.DROPOUT,
).to(cfg.DEVICE)

#Layerwise LR decay (LLRD) — kỹ thuật đặc trưng của DeBERTa 
# Các layer sâu hơn (gần output) được fine-tune với lr cao hơn
# Các layer nông hơn (gần input — embedding) lr nhỏ hơn → bảo toàn pretrained
LLRD_DECAY = 0.9   # mỗi layer nhân thêm 0.9 so với layer trên

def get_deberta_param_groups(model, base_lr, decay=LLRD_DECAY, head_lr=cfg.LR_HEAD):
    """
    Tạo param groups với LLRD:
      - classifier head          → head_lr
      - encoder layer N (cuối)   → base_lr
      - encoder layer N-1        → base_lr * decay
      - ...
      - embeddings               → base_lr * decay^N
    """
    # Classifier head
    param_groups = [{"params": list(model.classifier.parameters()) +
                               list(model.norm.parameters()),
                     "lr": head_lr, "weight_decay": 0.0}]

    # Encoder layers (từ trên xuống)
    num_layers = model.deberta.config.num_hidden_layers  # 12
    for layer_idx in range(num_layers - 1, -1, -1):
        lr_layer = base_lr * (decay ** (num_layers - 1 - layer_idx))
        params   = list(model.deberta.encoder.layer[layer_idx].parameters())
        param_groups.append({"params": params, "lr": lr_layer,
                              "weight_decay": cfg.WEIGHT_DECAY})

    # Embeddings
    emb_lr = base_lr * (decay ** num_layers)
    param_groups.append({"params": list(model.deberta.embeddings.parameters()),
                         "lr": emb_lr, "weight_decay": cfg.WEIGHT_DECAY})

    return param_groups

try:
    param_groups = get_deberta_param_groups(model, cfg.LR_DEBERTA)
    print(f"LLRD param groups: {len(param_groups)} groups")
except AttributeError:
    # Fallback nếu kiến trúc encoder khác
    print("[INFO] LLRD fallback: dùng 2 group đơn giản")
    param_groups = [
        {"params": model.deberta.parameters(),  "lr": cfg.LR_DEBERTA, "weight_decay": cfg.WEIGHT_DECAY},
        {"params": list(model.classifier.parameters()) + list(model.norm.parameters()),
         "lr": cfg.LR_HEAD, "weight_decay": 0.0},
    ]

optimizer = AdamW(param_groups, eps=1e-6)

# Class-weighted loss
class_counts  = np.bincount(train_labels)
class_weights = torch.tensor(1.0 / class_counts, dtype=torch.float)
class_weights = (class_weights / class_weights.sum() * cfg.NUM_CLASSES).to(cfg.DEVICE)
criterion     = nn.CrossEntropyLoss(weight=class_weights)

# Linear Warmup scheduler
total_steps  = (len(train_loader) // cfg.GRAD_ACCUM) * cfg.EPOCHS
warmup_steps = int(total_steps * cfg.WARMUP_RATIO)

def warmup_lambda(step):
    if step < warmup_steps:
        return float(step) / float(max(1, warmup_steps))
    return 1.0

warmup_scheduler = LambdaLR(optimizer, lr_lambda=warmup_lambda)
reduce_scheduler = ReduceLROnPlateau(optimizer, mode="max", factor=0.5, patience=2)

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Trainable parameters: {trainable:,}")

# 7. Train / Eval helpers
def train_epoch(model, loader, criterion, optimizer,
                warmup_sched, global_step, warmup_steps, grad_accum):
    model.train()
    total_loss = total_correct = total = 0
    optimizer.zero_grad()

    for step, batch in enumerate(loader):
        input_ids       = batch["input_ids"].to(cfg.DEVICE)
        attention_mask  = batch["attention_mask"].to(cfg.DEVICE)
        token_type_ids  = batch["token_type_ids"].to(cfg.DEVICE)
        labels          = batch["label"].to(cfg.DEVICE)

        logits, _ = model(input_ids, attention_mask, token_type_ids)
        loss      = criterion(logits, labels) / grad_accum
        loss.backward()

        if (step + 1) % grad_accum == 0 or (step + 1) == len(loader):
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            optimizer.zero_grad()
            global_step += 1
            if global_step <= warmup_steps:
                warmup_sched.step()

        total_loss    += loss.item() * grad_accum * input_ids.size(0)
        total_correct += (logits.argmax(1) == labels).sum().item()
        total         += input_ids.size(0)

    return total_loss / total, total_correct / total, global_step


def eval_epoch(model, loader, criterion):
    model.eval()
    total_loss = total_correct = total = 0
    all_preds, all_labels = [], []

    with torch.no_grad():
        for batch in loader:
            input_ids      = batch["input_ids"].to(cfg.DEVICE)
            attention_mask = batch["attention_mask"].to(cfg.DEVICE)
            token_type_ids = batch["token_type_ids"].to(cfg.DEVICE)
            labels         = batch["label"].to(cfg.DEVICE)

            logits, _ = model(input_ids, attention_mask, token_type_ids)
            loss      = criterion(logits, labels)

            total_loss    += loss.item() * input_ids.size(0)
            total_correct += (logits.argmax(1) == labels).sum().item()
            total         += input_ids.size(0)
            all_preds.extend(logits.argmax(1).cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    f1 = f1_score(all_labels, all_preds, average="macro")
    return total_loss / total, total_correct / total, f1


def predict_all(model, loader):
    model.eval()
    all_probs, all_preds, all_labels = [], [], []

    with torch.no_grad():
        for batch in loader:
            input_ids      = batch["input_ids"].to(cfg.DEVICE)
            attention_mask = batch["attention_mask"].to(cfg.DEVICE)
            token_type_ids = batch["token_type_ids"].to(cfg.DEVICE)
            labels         = batch["label"]

            logits, _ = model(input_ids, attention_mask, token_type_ids)
            probs     = torch.softmax(logits, dim=-1).cpu().numpy()
            preds     = logits.argmax(1).cpu().numpy()

            all_probs.extend(probs)
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())

    return np.array(all_probs), np.array(all_preds), np.array(all_labels)

# 8. Training loop với Early Stopping
history = {
    "train_loss": [], "train_acc": [],
    "val_loss":   [], "val_acc":   [], "val_f1": [],
}

best_val_f1  = -1
best_epoch   = 0
patience_cnt = 0
best_state   = None
global_step  = 0

print("\n" + "=" * 72)
print(f"{'Ep':>4} {'TrLoss':>9} {'TrAcc':>8} "
      f"{'ValLoss':>9} {'ValAcc':>8} {'ValF1':>8} {'LR_enc':>12}")
print("=" * 72)

for epoch in range(1, cfg.EPOCHS + 1):
    tr_loss, tr_acc, global_step = train_epoch(
        model, train_loader, criterion, optimizer,
        warmup_scheduler, global_step, warmup_steps, cfg.GRAD_ACCUM
    )
    val_loss, val_acc, val_f1 = eval_epoch(model, valid_loader, criterion)

    if global_step > warmup_steps:
        reduce_scheduler.step(val_f1)

    # LR của encoder layer cuối (group 1)
    current_lr = optimizer.param_groups[1]["lr"]

    history["train_loss"].append(tr_loss)
    history["train_acc"].append(tr_acc)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)
    history["val_f1"].append(val_f1)

    marker = " ★" if val_f1 > best_val_f1 else ""
    print(f"{epoch:>4} {tr_loss:>9.4f} {tr_acc:>8.4f} "
          f"{val_loss:>9.4f} {val_acc:>8.4f} {val_f1:>8.4f} "
          f"{current_lr:>12.2e}{marker}")

    if val_f1 > best_val_f1:
        best_val_f1  = val_f1
        best_epoch   = epoch
        patience_cnt = 0
        best_state   = {k: v.cpu().clone() for k, v in model.state_dict().items()}
    else:
        patience_cnt += 1
        if patience_cnt >= cfg.PATIENCE:
            print(f"\nEarly stopping tại epoch {epoch} "
                  f"(best epoch={best_epoch}, val F1={best_val_f1:.4f})")
            break

model.load_state_dict(best_state)
model.to(cfg.DEVICE)
print(f"\nĐã khôi phục best model từ epoch {best_epoch}.")

torch.save({
    "model_state": best_state,
    "config":      cfg,
    "best_epoch":  best_epoch,
    "best_val_f1": best_val_f1,
}, "mdeberta_best.pt")
print("Đã lưu model → mdeberta_best.pt")

# 9. Plot training curves
epochs_ran = list(range(1, len(history["train_loss"]) + 1))
fig, axes  = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(epochs_ran, history["train_loss"], marker="o", label="Train Loss")
axes[0].plot(epochs_ran, history["val_loss"],   marker="s", label="Val Loss")
axes[0].axvline(best_epoch, linestyle="--", color="red", alpha=0.5,
                label=f"Best epoch={best_epoch}")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Loss")
axes[0].set_title("Loss Curves"); axes[0].legend(); axes[0].grid(True)

axes[1].plot(epochs_ran, history["train_acc"], marker="o", label="Train Acc")
axes[1].plot(epochs_ran, history["val_acc"],   marker="s", label="Val Acc")
axes[1].axvline(best_epoch, linestyle="--", color="red", alpha=0.5)
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("Accuracy")
axes[1].set_title("Accuracy Curves"); axes[1].legend(); axes[1].grid(True)

axes[2].plot(epochs_ran, history["val_f1"], marker="^",
             color="green", label="Val F1 macro")
axes[2].axvline(best_epoch, linestyle="--", color="red", alpha=0.5,
                label=f"Best epoch={best_epoch}")
axes[2].set_xlabel("Epoch"); axes[2].set_ylabel("F1 Macro")
axes[2].set_title("Validation F1 Macro"); axes[2].legend(); axes[2].grid(True)

plt.suptitle("mDeBERTa-v3-base – Training History", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("mdeberta_training_curves.png", dpi=150)
plt.show()

# 10. Evaluate on test set
y_pred_probs, y_pred, y_true = predict_all(model, test_loader)

acc        = accuracy_score(y_true, y_pred)
prec_macro = precision_score(y_true, y_pred, average="macro")
rec_macro  = recall_score(y_true, y_pred, average="macro")
f1_macro   = f1_score(y_true, y_pred, average="macro")
y_true_bin = label_binarize(y_true, classes=[0, 1, 2])
auc_macro  = roc_auc_score(y_true_bin, y_pred_probs,
                            average="macro", multi_class="ovr")

print("\n" + "=" * 60)
print("FINAL TEST SET METRICS – mDeBERTa-v3-base + Teencode (4 cột)")
print("=" * 60)
print(f"{'Metric':<25} {'Value':>10}")
print("-" * 60)
print(f"{'Accuracy':<25} {acc:>10.4f}")
print(f"{'Precision (macro)':<25} {prec_macro:>10.4f}")
print(f"{'Recall (macro)':<25} {rec_macro:>10.4f}")
print(f"{'F1-score (macro)':<25} {f1_macro:>10.4f}")
print(f"{'AUC (macro)':<25} {auc_macro:>10.4f}")
print("=" * 60)
print("\nCLASSIFICATION REPORT")
print(classification_report(y_true, y_pred, target_names=CLASS_NAMES))

# 11. Confusion Matrix
cf = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(7, 6))
sn.heatmap(cf, annot=True, fmt="d", cmap="Blues",
           xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
           cbar_kws={"label": "Count"})
plt.xlabel("Predicted", fontsize=12); plt.ylabel("True", fontsize=12)
plt.title("Confusion Matrix – mDeBERTa-v3-base (Test Set)",
          fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig("mdeberta_confusion_matrix.png", dpi=150)
plt.show()

# 12. Class-wise bar chart
prec_cls = precision_score(y_true, y_pred, average=None)
rec_cls  = recall_score(y_true, y_pred, average=None)
f1_cls   = f1_score(y_true, y_pred, average=None)

x = np.arange(len(CLASS_NAMES)); w = 0.25
plt.figure(figsize=(9, 5))
plt.bar(x - w, prec_cls, w, label="Precision", alpha=0.85)
plt.bar(x,     rec_cls,  w, label="Recall",    alpha=0.85)
plt.bar(x + w, f1_cls,   w, label="F1-score",  alpha=0.85)
plt.xticks(x, CLASS_NAMES)
plt.ylim([0, 1.1]); plt.ylabel("Score")
plt.title("Class-wise Metrics – mDeBERTa-v3-base + Teencode (4 cột)",
          fontweight="bold")
plt.legend(); plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("mdeberta_classwise_metrics.png", dpi=150)
plt.show()

# 13. ROC Curves
fpr, tpr, roc_auc_d = {}, {}, {}
for i in range(3):
    fpr[i], tpr[i], _ = roc_curve(y_true_bin[:, i], y_pred_probs[:, i])
    roc_auc_d[i]      = auc(fpr[i], tpr[i])
fpr["macro"], tpr["macro"], _ = roc_curve(y_true_bin.ravel(), y_pred_probs.ravel())
roc_auc_d["macro"] = auc(fpr["macro"], tpr["macro"])

colors = ["#FF6B6B", "#4ECDC4", "#45B7D1"]
plt.figure(figsize=(9, 7))
for i in range(3):
    plt.plot(fpr[i], tpr[i], color=colors[i], lw=2,
             label=f"{CLASS_NAMES[i]} (AUC={roc_auc_d[i]:.4f})")
plt.plot(fpr["macro"], tpr["macro"], color="deeppink", linestyle="--", lw=2.5,
         label=f"macro-avg (AUC={roc_auc_d['macro']:.4f})")
plt.plot([0, 1], [0, 1], "k--", lw=1.5, alpha=0.5, label="Random")
plt.xlim([0, 1]); plt.ylim([0, 1.05])
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("ROC Curves – mDeBERTa-v3-base + Teencode (4 cột)", fontweight="bold")
plt.legend(loc="lower right"); plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("mdeberta_roc_curves.png", dpi=150)
plt.show()

# 14. Attention Heatmap — 5 mẫu test (layer cuối, head 0)
def plot_deberta_attention(model, tokenizer, texts, y_true, y_pred,
                           max_len, n=5, layer=-1, head=0):
    model.eval()
    fig, axes = plt.subplots(n, 1, figsize=(16, n * 3.5))
    if n == 1:
        axes = [axes]

    for i in range(n):
        enc = tokenizer(
            texts[i],
            max_length     = max_len,
            padding        = "max_length",
            truncation     = True,
            return_tensors = "pt",
        )
        input_ids      = enc["input_ids"].to(cfg.DEVICE)
        attention_mask = enc["attention_mask"].to(cfg.DEVICE)

        with torch.no_grad():
            outputs = model.deberta(
                input_ids         = input_ids,
                attention_mask    = attention_mask,
                output_attentions = True,
            )

        # mDeBERTa có thể trả None cho attentions nếu kiến trúc không support
        if outputs.attentions is None:
            print("[INFO] mDeBERTa không trả attention weights — bỏ qua heatmap.")
            plt.close()
            return

        attn   = outputs.attentions[layer][0, head].cpu().numpy()
        tokens = tokenizer.convert_ids_to_tokens(input_ids[0].cpu())
        seq_len = attention_mask[0].sum().item()
        tokens  = tokens[:seq_len]
        attn    = attn[:seq_len, :seq_len]

        true_lbl = ID2LABEL[y_true[i]]
        pred_lbl = ID2LABEL[y_pred[i]]
        mark     = "✓" if true_lbl == pred_lbl else "✗"

        ax = axes[i]
        im = ax.imshow(attn, cmap="Blues", aspect="auto")
        ax.set_xticks(range(len(tokens)))
        ax.set_xticklabels(tokens, rotation=45, ha="right", fontsize=7)
        ax.set_yticks(range(len(tokens)))
        ax.set_yticklabels(tokens, fontsize=7)
        ax.set_title(
            f"[{mark}] True: {true_lbl} | Pred: {pred_lbl}  —  {texts[i][:55]}",
            fontsize=9
        )
        plt.colorbar(im, ax=ax, fraction=0.02)

    plt.suptitle(f"mDeBERTa Attention Heatmap – Layer {layer}, Head {head}",
                 fontsize=13, fontweight="bold")
    plt.tight_layout()
    plt.savefig("mdeberta_attention_heatmap.png", dpi=150)
    plt.show()

plot_deberta_attention(model, tokenizer, test_texts, y_true, y_pred,
                       cfg.MAX_LENGTH, n=5)

# 15. Example predictions
print("\nEXAMPLE PREDICTIONS (first 10 test samples)")
print("-" * 70)
for i in range(min(10, len(test_texts))):
    orig     = test_df["sentence"].iloc[i]
    norm     = test_texts[i]
    true_lbl = ID2LABEL[y_true[i]]
    pred_lbl = ID2LABEL[y_pred[i]]
    probs    = y_pred_probs[i]
    mark     = "✓" if true_lbl == pred_lbl else "✗"
    print(f"[{mark}] Gốc  : {orig[:60]}")
    if orig.lower() != norm:
        print(f"     Chuẩn: {norm[:60]}")
    print(f"     True: {true_lbl:>8} | Pred: {pred_lbl:>8} | "
          f"neg={probs[0]:.3f} neu={probs[1]:.3f} pos={probs[2]:.3f}")
    print()

# Lưu kết quả để so sánh với BiLSTM và ViSoBERT
all_results = {}
all_results["mDeBERTa_v3"] = {
    "accuracy":        acc,
    "precision_macro": prec_macro,
    "recall_macro":    rec_macro,
    "f1_macro":        f1_macro,
    "auc_macro":       auc_macro,
}
print("Stored in all_results:", all_results)
```
