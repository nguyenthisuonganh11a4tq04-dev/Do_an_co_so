```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python

#Đọc file đã clean

df = pd.read_excel("data_cleaned.xlsx")

print("Số dòng:", len(df))
print("Các cột:", df.columns)

```

```python

#Kiểm tra có cột sentiment không

use_stratify = False

if 'sentiment' in df.columns:
    print("\nPhân bố label ban đầu:")
    print(df['sentiment'].value_counts())

    # XÓA label quá ít (<2 dòng)
    label_counts = df['sentiment'].value_counts()
    valid_labels = label_counts[label_counts >= 2].index

    df = df[df['sentiment'].isin(valid_labels)]

    print("\nSau khi lọc label hiếm:")
    print(df['sentiment'].value_counts())

    use_stratify = True

```

```python

#Chia train (70%) + temp (30%)

if use_stratify:
    train_df, temp_df = train_test_split(
        df,
        test_size=0.3,
        stratify=df['sentiment'],
        random_state=42
    )
else:
    train_df, temp_df = train_test_split(
        df,
        test_size=0.3,
        random_state=42
    )
```

```python

#Chia temp → val + test (KHÔNG stratify)
val_df, test_df = train_test_split(
    temp_df,
    test_size=2/3,
    random_state=42
)
```

```python

#Lưu file

train_df.to_excel("train.xlsx", index=False)
val_df.to_excel("val.xlsx", index=False)
test_df.to_excel("test.xlsx", index=False)

#In kết quả

print("\n===== KẾT QUẢ =====")
print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))
```
