```python
import pandas as pd
import re
```

```python
df = pd.read_excel("data post.xlsx")

# In tên cột để kiểm tra
print("Các cột trong file:")
print(df.columns)

# Lấy cột nội dung bài post
df = df[['text']] 
```

```python
# Làm sạch dữ liệu 
def clean_text(text):

    if pd.isna(text):
        return ""

    text = str(text)

    # xóa link
    text = re.sub(r'http\S+|www\S+', '', text)

    # xóa ký tự đặc biệt
    text = re.sub(r'[^a-zA-ZÀ-ỹ0-9\s]', ' ', text)

    # xóa khoảng trắng dư
    text = re.sub(r'\s+', ' ', text).strip()

    return text.lower()

df['sentence'] = df['text'].apply(clean_text)

```

```python
# Loại bỏ các bài quảng cáo
ads_keywords = [
    "bán","sale","ship","mua","inbox","ib",
    "tuyển","việc làm","part time","full time",
    "phòng trọ","cho thuê","sang nhượng","tips","góc"
]

def is_ads(text):
    for word in ads_keywords:
        if word in text:
            return True
    return False

# loại bài quảng cáo
df = df[~df['sentence'].apply(is_ads)]
```

```python
# Các từ khóa cảm xúc
positive_words = [
    "tốt","hay","vui","tuyệt","ok","ổn",
    "hài lòng","thích","giỏi","xuất sắc"
]

negative_words = [
    "tệ","chán","không","dở","kém",
    "bực","khó","tức","ghét"
]

def label_sentiment(text):

    for word in negative_words:
        if word in text:
            return "negative"

    for word in positive_words:
        if word in text:
            return "positive"

    return "neutral"

df['sentiment'] = df['sentence'].apply(label_sentiment)
```

```python

#Dataset cuối

df_final = df[['sentiment','sentence']]

# xóa dòng trống
df_final = df_final[df_final['sentence'] != ""]

```

```python
# Lưu file

df_final.to_excel("clean_labeled_posts2.xlsx", index=False)

print("Hoàn thành!")
print("Số dòng dữ liệu:", len(df_final))
```
