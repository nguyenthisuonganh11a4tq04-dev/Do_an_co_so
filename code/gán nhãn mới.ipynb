```python
import pandas as pd

df = pd.read_excel("C:/Users/user/OneDrive/Desktop/do an co so/data/data_cleaned.xlsx")

print("Số dòng:", len(df))
print("Các cột:", df.columns)
```

```python
def classify_sentence(text):
    text = str(text).lower()

    # ===== INQUIRY =====
    if "?" in text or any(w in text for w in ["không", "sao", "bao nhiêu", "khi nào", "ở đâu"]):
        
        if any(w in text for w in ["đăng ký", "môn", "lịch học", "thi", "điểm"]):
            return "Inquiry_Academic"
        elif any(w in text for w in ["học phí", "đóng tiền", "học bổng"]):
            return "Inquiry_Finance"
        elif any(w in text for w in ["web", "app", "lỗi"]):
            return "Inquiry_System"
        elif any(w in text for w in ["nghỉ", "bảo lưu", "giấy", "quy định"]):
            return "Inquiry_Policy"
        elif any(w in text for w in ["wifi", "phòng", "bãi xe"]):
            return "Inquiry_Facility"
        else:
            return "Inquiry_Other"

    # ===== COMPLAINT =====
    elif any(w in text for w in ["tệ", "chán", "lỗi", "lag", "bực", "kém", "nóng"]):
        
        if any(w in text for w in ["web", "đăng ký", "app"]):
            return "Complaint_System"
        elif any(w in text for w in ["thầy", "cô", "giảng viên"]):
            return "Complaint_Teaching"
        elif any(w in text for w in ["wifi", "phòng", "nóng"]):
            return "Complaint_Facility"
        elif any(w in text for w in ["trường", "dịch vụ"]):
            return "Complaint_Service"
        elif any(w in text for w in ["học nhiều", "lịch"]):
            return "Complaint_Workload"
        else:
            return "Complaint_Other"

    # ===== PRAISE =====
    elif any(w in text for w in ["tốt", "hay", "ok", "đẹp", "xịn", "tuyệt"]):
        
        if any(w in text for w in ["thầy", "cô", "giảng viên"]):
            return "Praise_Teaching"
        elif any(w in text for w in ["trường", "phòng"]):
            return "Praise_Facility"
        elif any(w in text for w in ["hỗ trợ", "nhân viên"]):
            return "Praise_Service"
        else:
            return "Praise_Other"

    # ===== STATEMENT =====
    else:
        if any(w in text for w in ["hôm nay", "đang", "vừa"]):
            return "Statement_Status"
        elif any(w in text for w in ["đã đăng ký", "đóng học phí"]):
            return "Statement_Academic"
        elif any(w in text for w in ["mai", "ngày mai", "có lịch"]):
            return "Statement_Announcement"
        else:
            return "Statement_Other"
```

```python
#TẠO CỘT MỚI

df['aspect_label'] = df['sentence'].apply(classify_sentence)
```
